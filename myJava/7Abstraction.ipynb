# Abstraction in Java (Abstract Classes)
**Abstraction** is the process of hiding internal implementation details and showing only the essential features to the user.

### Key Rules of Abstract Classes:
1. Declared using the `abstract` keyword.
2. **Cannot be instantiated** directly (`new BaseClass()` is invalid).
3. Can contain both **abstract methods** (no body) and **concrete methods** (with body).
4. Can have constructors, fields, and `static` methods.
5. Any subclass extending an abstract class must implement all inherited abstract methods (unless the subclass is also abstract).


---
## Abstract Class Example
- `Shape`: Abstract superclass defining an abstract method `area()` and a concrete method `display()`.
- `Circle` & `Rectangle`: Concrete subclasses providing specific `area()` implementations.


```java
// Abstract class
abstract class Shape {
    String color;

    // Constructor in abstract class
    Shape(String color) {
        this.color = color;
    }

    // Abstract method (no implementation body)
    abstract double area();

    // Concrete method (with implementation)
    void display() {
        System.out.println("Color: " + this.color + ", Area: " + area());
    }
}

// Concrete subclass 1
class Circle extends Shape {
    double radius;

    Circle(String color, double radius) {
        super(color);
        this.radius = radius;
    }

    @Override
    double area() {
        return Math.PI * radius * radius;
    }
}

// Concrete subclass 2
class Rectangle extends Shape {
    double width, height;

    Rectangle(String color, double width, double height) {
        super(color);
        this.width = width;
        this.height = height;
    }

    @Override
    double area() {
        return width * height;
    }
}

// Testing Abstraction
Shape s1 = new Circle("Red", 5.0);
Shape s2 = new Rectangle("Blue", 4.0, 6.0);

s1.display();
s2.display();

// Direct instantiation is blocked:
// Shape s = new Shape("Green"); // Compile-time error: Shape is abstract; cannot be instantiated
```

## 3. Inner Classes in Java
Java supports 4 types of nested/inner classes:
1. **Regular (Member) Inner Class**: Defined non-statically inside a class. Has full access to outer class instance fields.
2. **Method-Local Inner Class**: Defined inside a method body; scope is restricted to that method.
3. **Anonymous Inner Class**: An unnamed class declared and instantiated in a single expression (often used to implement interfaces on the fly).
4. **Static Nested Class**: Declared with `static`; cannot directly access outer class non-static fields without an instance.


```java
// 1. Outer Class with Member Inner and Static Nested
class OuterClass {
    private String secretMsg = "Confidential Data";
    static String staticInfo = "Global Server Info";

    // Regular Inner Class
    class InnerClass {
        void printOuter() {
            System.out.println("Inner class accessing outer: " + secretMsg);
        }
    }

    // Static Nested Class
    static class StaticNestedClass {
        void printStatic() {
            System.out.println("Static nested accessing: " + staticInfo);
        }
    }
}

// 2. Instantiating Member vs Static Nested
OuterClass outer = new OuterClass();
OuterClass.InnerClass inner = outer.new InnerClass();
inner.printOuter();

OuterClass.StaticNestedClass staticNested = new OuterClass.StaticNestedClass();
staticNested.printStatic();

// 3. Anonymous Inner Class implementing a quick interface
interface GreetingService {
    void greet(String name);
}

// Implementing anonymously on-the-fly
GreetingService greeter = new GreetingService() {
    @Override
    public void greet(String name) {
        System.out.println("Hello, " + name + " (from Anonymous Inner Class!)");
    }
};
greeter.greet("Swaroop");
```

### 4. Method-Local Inner Class Demonstration (Topic 53)
- Defined completely inside the body of a method.
- Scope is strictly restricted to that method.
- Can access local variables and parameters of the enclosing method only if they are **`final` or effectively final** (not modified after assignment).


```java
class BankTransactionService {
    private String branchCode = "PUNE-ACTS-01";

    public void processPayment(double amount) {
        // Local variable (effectively final)
        String transactionId = "TXN-" + System.currentTimeMillis();

        // Method-Local Inner Class defined inside method scope
        class Validator {
            boolean isValid() {
                return amount > 0 && amount <= 100000;
            }

            void logReceipt() {
                // Accesses outer class private field and method local variable
                System.out.println("[Method-Local Inner Class] Branch: " + branchCode);
                System.out.println("[Method-Local Inner Class] ID: " + transactionId + " | Amount: Rs. " + amount);
            }
        }

        // Instantiate and execute inside method
        Validator validator = new Validator();
        if (validator.isValid()) {
            System.out.println("Payment of Rs. " + amount + " successfully validated!");
            validator.logReceipt();
        } else {
            System.out.println("Payment validation failed for Rs. " + amount);
        }
    }
}

BankTransactionService service = new BankTransactionService();
service.processPayment(25000.0);
```
