# Strings in Java
Java provides **3 core types of string classes** to handle character sequences based on mutability and thread-safety:

### The 3 Core String Classes in Java:
| Class | Mutability | Thread-Safe? | Storage Area | Performance / Best Use |
| :--- | :--- | :--- | :--- | :--- |
| **`String`** | **Immutable** (cannot be changed) | ✅ Yes (inherently) | String Constant Pool (SCP) / Heap | Read-only text, keys, constants |
| **`StringBuilder`** | **Mutable** (modifiable in-place) | ❌ No (non-synchronized) | Heap | Fast single-threaded modifications |
| **`StringBuffer`** | **Mutable** (modifiable in-place) | ✅ Yes (synchronized methods) | Heap | Multi-threaded string modifications |

> *Note on Modern Java (Java 15+)*: Includes **Text Blocks** (`"""..."""`) for multiline strings.


---
## 1. `String` Class (Immutable)
- Created via **String Literal** (stored in String Constant Pool) or with **`new String()`** (stored in heap).
- **Immutability**: Any modification creates a new object rather than altering the existing one.


```java
// Literal vs new Object
String s1 = "Java";                 // stored in String Constant Pool
String s2 = "Java";                 // reuses s1 from pool
String s3 = new String("Java");      // new object created in Heap

System.out.println("s1 == s2 (same reference?): " + (s1 == s2));      // true
System.out.println("s1 == s3 (same reference?): " + (s1 == s3));      // false
System.out.println("s1.equals(s3) (same content?): " + s1.equals(s3)); // true

// Immutability demonstration
String original = "Hello";
original.concat(" World"); // return value ignored
System.out.println("Original unchanged: " + original); // still 'Hello'

original = original.concat(" World"); // explicitly reassigning
System.out.println("After reassigning: " + original);
```

---
## Key `String` Methods
- `length()`, `charAt(index)`, `substring(start, end)`
- `toUpperCase()`, `toLowerCase()`, `trim()`, `strip()`
- `contains()`, `startsWith()`, `endsWith()`
- `replace()`, `split()`, `toCharArray()`, `indexOf()`


```java
String text = "  Learn Java Programming  ";

System.out.println("length()       : " + text.length());
System.out.println("charAt(4)      : " + text.charAt(4));
System.out.println("trim()         : '" + text.trim() + "'");
System.out.println("toUpperCase()  : " + text.toUpperCase());
System.out.println("substring(2, 7): " + text.substring(2, 7));
System.out.println("contains('Java'): " + text.contains("Java"));
System.out.println("replace('a','*'): " + text.replace('a', '*'));

// split() example
String csv = "Red,Green,Blue";
String[] colors = csv.split(",");
System.out.println("split() length : " + colors.length + ", first: " + colors[0]);
```

---
## 2. `StringBuilder` Class (Mutable, Fast, Non-Synchronized)
- Modifies text directly without creating new garbage objects.
- Best choice for string manipulation in single-threaded code.
- Key methods: `append()`, `insert()`, `replace()`, `delete()`, `reverse()`.


```java
StringBuilder sb = new StringBuilder("Hello");

// In-place append
sb.append(" World");
System.out.println("append() : " + sb);

// In-place insert
sb.insert(5, ",");
System.out.println("insert() : " + sb);

// In-place replace and delete
sb.replace(0, 5, "Hi");
System.out.println("replace(): " + sb);
sb.delete(2, 3); // remove comma
System.out.println("delete() : " + sb);

// In-place reverse
sb.reverse();
System.out.println("reverse(): " + sb);
```

---
## 3. `StringBuffer` Class (Mutable, Thread-Safe, Synchronized)
- Same methods as `StringBuilder`, but its methods are `synchronized`.
- Safe to share across multiple concurrent threads, but has minor locking overhead compared to `StringBuilder`.


```java
StringBuffer sbuf = new StringBuffer("CDAC");

sbuf.append(" Java");
sbuf.append(" 2026");
System.out.println("StringBuffer: " + sbuf);
System.out.println("Capacity    : " + sbuf.capacity());

sbuf.reverse();
System.out.println("Reversed    : " + sbuf);
```

---
## 4. Modern Java: Text Blocks (`"""..."""`)
Introduced in modern Java for formatted multiline strings without manual `\n` or escape sequences.


```java
String jsonBlock = """
    {
        "name": "Swaroop",
        "course": "Java",
        "active": true
    }
    """;

System.out.println("--- Multiline Text Block ---");
System.out.println(jsonBlock);
```
