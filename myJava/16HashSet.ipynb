# HashSet in Java (`java.util.HashSet`)
A **HashSet** is a collection that stores elements using a **hash table** (internally backed by a `HashMap`). It implements the `Set` interface.

### Key Characteristics:
- **No Duplicates**: Duplicate elements are ignored (`add()` returns `false`).
- **Unordered**: Does not maintain any insertion or sorted order.
- **Null Values**: Allows at most **one** `null` element.
- **Performance**: $O(1)$ constant time for basic operations (`add`, `remove`, `contains`) assuming a good hash function.

### Complete Methods Reference:
| Method | Description |
| :--- | :--- |
| `add(e)` | Adds element; returns `true` if added, `false` if duplicate |
| `addAll(collection)` | Adds all elements from another collection (Union) |
| `contains(o)` | Returns `true` if element exists |
| `containsAll(collection)` | Checks if all elements exist (Subset test) |
| `remove(o)` | Removes the specified element |
| `removeAll(collection)` | Removes matching elements (Difference / Complement) |
| `retainAll(collection)` | Keeps only elements present in both sets (Intersection) |
| `removeIf(predicate)` | Removes elements satisfying a condition |
| `size()` | Returns total element count |
| `isEmpty()` | Checks if set contains no elements |
| `toArray()` | Converts set to an array |
| `clone()` | Creates a shallow copy of the set |
| `clear()` | Removes all elements |


```java
import java.util.HashSet;
import java.util.Arrays;
import java.util.Iterator;

// 1. Creation and add() with duplicate check
HashSet<String> set = new HashSet<>();
System.out.println("add('Java')     : " + set.add("Java"));
System.out.println("add('Python')   : " + set.add("Python"));
System.out.println("add('C++')      : " + set.add("C++"));
System.out.println("add('Java') dup : " + set.add("Java")); // returns false
set.add(null); // one null allowed
System.out.println("HashSet: " + set);

// 2. Query methods
System.out.println("contains('Python'): " + set.contains("Python"));
System.out.println("size(): " + set.size() + ", isEmpty(): " + set.isEmpty());

// 3. Removal
set.remove(null);
set.remove("C++");
set.removeIf(lang -> lang.startsWith("P")); // removes 'Python'
System.out.println("After removals: " + set);

// 4. Mathematical Set Operations (Union, Intersection, Difference)
HashSet<Integer> setA = new HashSet<>(Arrays.asList(1, 2, 3, 4, 5));
HashSet<Integer> setB = new HashSet<>(Arrays.asList(4, 5, 6, 7, 8));

// Union (addAll)
HashSet<Integer> union = new HashSet<>(setA);
union.addAll(setB);
System.out.println("\nUnion (setA U setB)       : " + union);

// Intersection (retainAll)
HashSet<Integer> intersection = new HashSet<>(setA);
intersection.retainAll(setB);
System.out.println("Intersection (setA ∩ setB): " + intersection);

// Difference (removeAll)
HashSet<Integer> difference = new HashSet<>(setA);
difference.removeAll(setB);
System.out.println("Difference (setA - setB)  : " + difference);

// 5. toArray and clone
Object[] arr = union.toArray();
System.out.println("\nArray conversion length: " + arr.length);
@SuppressWarnings("unchecked")
HashSet<Integer> cloned = (HashSet<Integer>) union.clone();
System.out.println("Cloned Set: " + cloned);

// 6. Iterating through HashSet
System.out.print("Iterating: ");
for (int val : setA) {
    System.out.print(val + " ");
}
System.out.println();

// 7. Clear
setA.clear();
System.out.println("After clear(), isEmpty(): " + setA.isEmpty());
```

## 2. `LinkedHashSet` and `TreeSet` (The Set Hierarchy)

| Feature | `HashSet` | `LinkedHashSet` | `TreeSet` |
| :--- | :--- | :--- | :--- |
| **Order** | Unordered | **Maintains Insertion Order** | **Sorted Natural Order** |
| **Backing Structure** | Hash Table | Hash Table + Doubly-Linked List | Red-Black Balanced Binary Search Tree |
| **Time Complexity** | $O(1)$ | $O(1)$ | $O(\log n)$ |
| **Null Allowed** | Yes (1 null) | Yes (1 null) | ❌ No nulls (throws `NullPointerException`) |


```java
import java.util.HashSet;
import java.util.LinkedHashSet;
import java.util.TreeSet;
import java.util.Arrays;

List<Integer> sample = Arrays.asList(40, 10, 50, 20, 30);

// 1. HashSet: No guaranteed order
HashSet<Integer> hashSet = new HashSet<>(sample);
System.out.println("HashSet (Unordered)      : " + hashSet);

// 2. LinkedHashSet: Preserves insertion order
LinkedHashSet<Integer> linkedSet = new LinkedHashSet<>(sample);
System.out.println("LinkedHashSet (Insertion): " + linkedSet);

// 3. TreeSet: Automatically sorted ascending
TreeSet<Integer> treeSet = new TreeSet<>(sample);
System.out.println("TreeSet (Sorted Ascending): " + treeSet);
System.out.println("TreeSet first() element  : " + treeSet.first());
System.out.println("TreeSet last() element   : " + treeSet.last());
```

## 3. Hashing Internals & the `equals()` and `hashCode()` Contract
When you insert a key into a `HashSet` or `HashMap`:
1. `hashCode()` is called to calculate the initial integer hash value.
2. The JVM calculates bucket index: `index = (n - 1) & hash`.
3. If collision occurs in that bucket, `equals()` is called on existing keys in that bucket:
   - If `equals()` returns `true`, the value is updated (no duplicate key).
   - If `equals()` returns `false`, a new node is linked.
4. **Java 8+ Bucket Treeification**: When bucket size exceeds 8 nodes and array capacity $\ge 64$, the bucket linked list converts into a **Red-Black Tree**, improving worst-case search from $O(n)$ to $O(\log n)$!

### The Official Contract:
- If `a.equals(b) == true`, then `a.hashCode() == b.hashCode()` **MUST** always be true.
- If `a.hashCode() == b.hashCode()`, `a.equals(b)` may or may not be true (hash collision).


```java
import java.util.Objects;
import java.util.HashSet;

class StudentKey {
    int rollNo;
    String name;

    StudentKey(int rollNo, String name) {
        this.rollNo = rollNo;
        this.name = name;
    }

    // Overriding equals and hashCode contract
    @Override
    public boolean equals(Object o) {
        if (this == o) return true;
        if (o == null || getClass() != o.getClass()) return false;
        StudentKey student = (StudentKey) o;
        return rollNo == student.rollNo && Objects.equals(name, student.name);
    }

    @Override
    public int hashCode() {
        return Objects.hash(rollNo, name);
    }

    @Override
    public String toString() {
        return rollNo + ":" + name;
    }
}

HashSet<StudentKey> studentSet = new HashSet<>();
studentSet.add(new StudentKey(1, "Swaroop"));
studentSet.add(new StudentKey(1, "Swaroop")); // recognized as duplicate because equals/hashCode overridden!

System.out.println("Set size (Expected 1): " + studentSet.size());
System.out.println("Set content: " + studentSet);
```

## 4. Load Factor, Initial Capacity, Rehashing & Resizing (Topics 92 & 93)

### A. Initial Capacity and Load Factor (Topic 92)
- **Initial Capacity (`16`)**: The number of buckets created in the hash table when instantiated with the default constructor (`new HashSet<>()`).
- **Load Factor (`0.75f`)**: A measure of how full the hash table is allowed to get before its capacity is automatically increased. Default is `0.75` (75% full).
- **Threshold Calculation**:
  $$\text{Threshold} = \text{Capacity} \times \text{Load Factor} = 16 \times 0.75 = 12$$
  *When the 13th element is inserted, the internal bucket array automatically triggers resizing!*

### B. Resizing and Rehashing Mechanics (Topic 93)
- **Resizing**: The capacity of the bucket array doubles ($2\times$): $16 \rightarrow 32 \rightarrow 64 \rightarrow 128 \dots$
- **Rehashing**: Because bucket indices are calculated as `index = (capacity - 1) & hash`, doubling the capacity changes the bitmask `(capacity - 1)`. Therefore, **every existing entry must have its bucket index recalculated** and moved to its new bucket.
- **Performance Note**: Rehashing is an $O(n)$ expensive operation. If the expected element count is known upfront, initialize the set with adequate capacity (`new HashSet<>(expectedSize / 0.75f + 1)`) to avoid costly rehashing cycles.


```java
import java.util.HashSet;

// Demonstrating custom Initial Capacity and Load Factor tuning
// Constructor: HashSet(int initialCapacity, float loadFactor)
int initialCapacity = 32;
float loadFactor = 0.80f; // 80% threshold

HashSet<String> optimizedSet = new HashSet<>(initialCapacity, loadFactor);
optimizedSet.add("Alpha");
optimizedSet.add("Beta");
optimizedSet.add("Gamma");

System.out.println("Tuned HashSet created with capacity 32 & load factor 0.80.");
System.out.println("Calculated Threshold before rehashing: " + (int)(32 * 0.80));
System.out.println("Current elements: " + optimizedSet);
```
