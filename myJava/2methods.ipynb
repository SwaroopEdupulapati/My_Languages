# Java Methods: All 8 Combinations of Method Types
A **method** is a collection of statements grouped together to execute a specific task, promote code reusability, and modularize programs.

Methods in Java vary across three primary binary dimensions:
1. **Static vs. Non-Static (Instance)**: Does the method belong to the class itself, or to individual object instances?
2. **Parameterized vs. Non-Parameterized**: Does the method accept arguments or take no inputs `()`?
3. **With Return Value vs. Without Return Value (`void`)**: Does the method return data to the caller or simply execute an action?

---

### Complete 8-Combination Matrix (2 × 2 × 2 = 8 Types):

| # | Modifier | Parameter Type | Return Type | Method Signature Pattern | Invocation Mechanism |
| :-: | :--- | :--- | :--- | :--- | :--- |
| **1** | **Static** | **Non-Parameterized** | **Without Return (`void`)** | `static void method()` | `ClassName.method()` |
| **2** | **Static** | **Parameterized** | **Without Return (`void`)** | `static void method(params)` | `ClassName.method(args)` |
| **3** | **Static** | **Non-Parameterized** | **With Return** | `static ReturnType method()` | `var = ClassName.method()` |
| **4** | **Static** | **Parameterized** | **With Return** | `static ReturnType method(params)` | `var = ClassName.method(args)` |
| **5** | **Non-Static** | **Non-Parameterized** | **Without Return (`void`)** | `void method()` | `obj.method()` |
| **6** | **Non-Static** | **Parameterized** | **Without Return (`void`)** | `void method(params)` | `obj.method(args)` |
| **7** | **Non-Static** | **Non-Parameterized** | **With Return** | `ReturnType method()` | `var = obj.method()` |
| **8** | **Non-Static** | **Parameterized** | **With Return** | `ReturnType method(params)` | `var = obj.method(args)` |

> Each combination is explained below with a dedicated breakdown, code definition, and invocation.

## PART 1: STATIC METHODS (Combinations 1 to 4)
- Declared using the `static` keyword.
- Belong to the **class** rather than any specific object.
- Can be invoked directly using the class name (`ClassName.methodName()`) without allocating memory via `new`.
- **Constraint**: Cannot directly access non-static instance fields or use `this` / `super`.


### Type 1: Static, Non-Parameterized, Without Return Value (`void`)
- **Nature**: Static method that takes no arguments `()` and returns nothing (`void`).
- **Use case**: Displaying general announcements, banners, utility headers, or system info.


```java
class StaticDemo1 {
    // 1. Static + Non-Parameterized + Without Return Value (void)
    public static void printWelcomeBanner() {
        System.out.println("===============================================");
        System.out.println(" Welcome to CDAC Java Methods Certification!  ");
        System.out.println("===============================================");
    }
}

// Invoked directly via ClassName without creating an object:
StaticDemo1.printWelcomeBanner();
```

### Type 2: Static, Parameterized, Without Return Value (`void`)
- **Nature**: Static method that accepts input arguments and performs an operation/logging without returning a value.
- **Use case**: Printing customized reports, processing log statements with parameters.


```java
class StaticDemo2 {
    // 2. Static + Parameterized + Without Return Value (void)
    public static void printStudentReport(String studentName, int rollNo, double marks) {
        System.out.println("--- Student Grade Card ---");
        System.out.println("Name    : " + studentName);
        System.out.println("Roll No : " + rollNo);
        System.out.println("Marks   : " + marks + "%");
        System.out.println("Status  : " + (marks >= 50.0 ? "PASSED" : "FAILED"));
    }
}

// Invocation with arguments passed directly to the class method:
StaticDemo2.printStudentReport("Swaroop", 101, 88.5);
```

### Type 3: Static, Non-Parameterized, With Return Value
- **Nature**: Static method that takes no arguments `()`, computes or retrieves an intrinsic value, and returns it to the caller.
- **Use case**: Fetching default global settings, system configuration, current server status, or standard constants.


```java
class StaticDemo3 {
    // 3. Static + Non-Parameterized + With Return Value
    public static double getDefaultTaxRate() {
        double baseGstRate = 0.18; // 18%
        return baseGstRate;
    }
    
    public static String getAppVersion() {
        return "v2.5.0-LTS";
    }
}

// Invocation: Caller receives and stores the returned value
double tax = StaticDemo3.getDefaultTaxRate();
String version = StaticDemo3.getAppVersion();

System.out.println("Application Version : " + version);
System.out.println("Default Tax Rate    : " + (tax * 100) + "%");
```

### Type 4: Static, Parameterized, With Return Value
- **Nature**: Pure functional static utility: receives input arguments, computes a result, and returns the output.
- **Use case**: Mathematical operations, formatting utilities, algorithmic evaluations (e.g. `Math.pow()`, `Math.max()`).


```java
class StaticDemo4 {
    // 4. Static + Parameterized + With Return Value
    public static double calculateSimpleInterest(double principal, double rate, int years) {
        double interest = (principal * rate * years) / 100.0;
        return interest;
    }
    
    public static boolean isEligibleToVote(int age) {
        return age >= 18;
    }
}

// Invocation: Pass parameters and receive returned value
double interestEarned = StaticDemo4.calculateSimpleInterest(25000.0, 7.5, 3);
boolean votingStatus = StaticDemo4.isEligibleToVote(21);

System.out.printf("Calculated Simple Interest : Rs. %.2f%n", interestEarned);
System.out.println("Eligible to Vote (Age 21)   : " + votingStatus);
```

## PART 2: NON-STATIC (INSTANCE) METHODS (Combinations 5 to 8)
- Declared **without** the `static` keyword.
- Belong to individual **object instances** of the class.
- **Require object instantiation** using the `new` keyword (`MyClass obj = new MyClass();`).
- Can freely access object instance state via the `this` reference as well as static class fields.


### Type 5: Non-Static, Non-Parameterized, Without Return Value (`void`)
- **Nature**: Instance method that operates on the caller object's own internal fields with no input parameters and no return value.
- **Use case**: Displaying object properties, triggering state updates or resets on a specific instance.


```java
class Car {
    String brand = "Tesla";
    String model = "Model 3";
    int speed = 0;
    
    // 5. Non-Static + Non-Parameterized + Without Return Value (void)
    public void displayCarDetails() {
        System.out.println("--- Car Specification ---");
        System.out.println("Brand        : " + this.brand);
        System.out.println("Model        : " + this.model);
        System.out.println("Current Speed: " + this.speed + " km/h");
    }
}

// Invocation: Requires creating an instance with 'new'
Car myCar = new Car();
myCar.displayCarDetails();
```

### Type 6: Non-Static, Parameterized, Without Return Value (`void`)
- **Nature**: Instance method that accepts input arguments to alter or act upon the specific object's internal state without returning a value.
- **Use case**: Setter methods, depositing money into an account, accelerating a vehicle.


```java
class BankAccount {
    String accountHolder;
    double balance;
    
    public BankAccount(String holder, double initialDeposit) {
        this.accountHolder = holder;
        this.balance = initialDeposit;
    }
    
    // 6. Non-Static + Parameterized + Without Return Value (void)
    public void deposit(double amount) {
        if (amount > 0) {
            this.balance += amount;
            System.out.printf("Successfully deposited Rs. %.2f for %s. New Balance: Rs. %.2f%n",
                              amount, this.accountHolder, this.balance);
        } else {
            System.out.println("Invalid deposit amount!");
        }
    }
}

// Invocation on specific object instance:
BankAccount acc = new BankAccount("Swaroop", 10000.0);
acc.deposit(3500.0);
```

### Type 7: Non-Static, Non-Parameterized, With Return Value
- **Nature**: Instance method that takes no arguments `()`, inspects or computes the object's instance state, and returns a value to the caller.
- **Use case**: Getter methods, querying current account balance, checking if an object's buffer is empty.


```java
class Student {
    String name;
    double score;
    
    public Student(String name, double score) {
        this.name = name;
        this.score = score;
    }
    
    // 7. Non-Static + Non-Parameterized + With Return Value
    public char calculateGrade() {
        if (this.score >= 90.0) return 'A';
        if (this.score >= 80.0) return 'B';
        if (this.score >= 70.0) return 'C';
        if (this.score >= 60.0) return 'D';
        return 'F';
    }
    
    public double getScore() {
        return this.score;
    }
}

// Invocation: Object method call returning computed values
Student s1 = new Student("Aarav", 86.5);
char studentGrade = s1.calculateGrade();
double currentScore = s1.getScore();

System.out.println("Student: " + s1.name + " | Score: " + currentScore + " | Grade: " + studentGrade);
```

### Type 8: Non-Static, Parameterized, With Return Value
- **Nature**: Instance method that takes parameters, combines them with the object's instance state, performs logic, and returns a computed result.
- **Use case**: Applying discounts or withdrawal transactions against an account balance, calculating compound interest with custom rates.


```java
class ECommerceCart {
    double cartTotal;
    
    public ECommerceCart(double initialTotal) {
        this.cartTotal = initialTotal;
    }
    
    // 8. Non-Static + Parameterized + With Return Value
    public double applyPromoCode(String promoCode, double discountPercentage) {
        if (promoCode != null && !promoCode.isEmpty()) {
            double discountAmount = (this.cartTotal * discountPercentage) / 100.0;
            double finalPayable = this.cartTotal - discountAmount;
            System.out.printf("Promo '%s' applied: Rs. %.2f discount on Rs. %.2f%n", 
                              promoCode, discountAmount, this.cartTotal);
            return finalPayable;
        }
        return this.cartTotal;
    }
}

// Invocation: Pass parameters into instance method and receive final result
ECommerceCart cart = new ECommerceCart(4500.0);
double payableAmount = cart.applyPromoCode("DIWALI50", 15.0);
System.out.printf("Final Amount to Pay: Rs. %.2f%n", payableAmount);
```

## Summary Comparison & Cheat Sheet

| # | Method Type | Key Identifier | Calling Syntax | Access to `this`? | Memory Allocation |
| :-: | :--- | :--- | :--- | :-: | :--- |
| **1** | Static, Non-Param, Void | `static void f()` | `Class.f()` | ❌ No | Class Area (Metaspace) |
| **2** | Static, Param, Void | `static void f(x)` | `Class.f(val)` | ❌ No | Class Area (Metaspace) |
| **3** | Static, Non-Param, Return | `static T f()` | `res = Class.f()` | ❌ No | Class Area (Metaspace) |
| **4** | Static, Param, Return | `static T f(x)` | `res = Class.f(val)` | ❌ No | Class Area (Metaspace) |
| **5** | Non-Static, Non-Param, Void | `void f()` | `obj.f()` | ✅ Yes | Heap (via Object Instance) |
| **6** | Non-Static, Param, Void | `void f(x)` | `obj.f(val)` | ✅ Yes | Heap (via Object Instance) |
| **7** | Non-Static, Non-Param, Return | `T f()` | `res = obj.f()` | ✅ Yes | Heap (via Object Instance) |
| **8** | Non-Static, Param, Return | `T f(x)` | `res = obj.f(val)` | ✅ Yes | Heap (via Object Instance) |

## All 8 Methods Together in a Single Unified Class
Here is an all-in-one class demonstration showing all 8 method types implemented side-by-side.


```java
class UnifiedMethodCatalog {
    // Instance variable
    int instanceVal = 50;
    // Static variable
    static int staticVal = 100;
    
    // 1. Static + Non-Parameterized + Void
    public static void m1_staticNonParamVoid() {
        System.out.println("1. [Static, Non-Param, Void] Called.");
    }
    
    // 2. Static + Parameterized + Void
    public static void m2_staticParamVoid(String msg) {
        System.out.println("2. [Static, Param, Void] Message: " + msg);
    }
    
    // 3. Static + Non-Parameterized + Return
    public static int m3_staticNonParamReturn() {
        return staticVal * 2;
    }
    
    // 4. Static + Parameterized + Return
    public static int m4_staticParamReturn(int multiplier) {
        return staticVal * multiplier;
    }
    
    // 5. Non-Static + Non-Parameterized + Void
    public void m5_instanceNonParamVoid() {
        System.out.println("5. [Instance, Non-Param, Void] instanceVal: " + this.instanceVal);
    }
    
    // 6. Non-Static + Parameterized + Void
    public void m6_instanceParamVoid(int addition) {
        this.instanceVal += addition;
        System.out.println("6. [Instance, Param, Void] New instanceVal: " + this.instanceVal);
    }
    
    // 7. Non-Static + Non-Parameterized + Return
    public int m7_instanceNonParamReturn() {
        return this.instanceVal;
    }
    
    // 8. Non-Static + Parameterized + Return
    public int m8_instanceParamReturn(int factor) {
        return this.instanceVal * factor;
    }
}

System.out.println("--- Calling Static Methods (1 to 4) ---");
UnifiedMethodCatalog.m1_staticNonParamVoid();
UnifiedMethodCatalog.m2_staticParamVoid("Hello Java");
System.out.println("3. Result: " + UnifiedMethodCatalog.m3_staticNonParamReturn());
System.out.println("4. Result: " + UnifiedMethodCatalog.m4_staticParamReturn(3));

System.out.println("\n--- Calling Non-Static / Instance Methods (5 to 8) ---");
UnifiedMethodCatalog catalogObj = new UnifiedMethodCatalog();
catalogObj.m5_instanceNonParamVoid();
catalogObj.m6_instanceParamVoid(25);
System.out.println("7. Result: " + catalogObj.m7_instanceNonParamReturn());
System.out.println("8. Result: " + catalogObj.m8_instanceParamReturn(2));
```
