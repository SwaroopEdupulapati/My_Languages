# Interfaces in Java
An **interface** is a blueprint of a class used to achieve **100% full abstraction** and support **multiple inheritance**.

### Key Rules of Interfaces:
1. Declared using the `interface` keyword and implemented using `implements`.
2. **Fields**: Automatically `public static final` (constants).
3. **Methods**: By default `public abstract` (prior to Java 8).
4. **Java 8+ features**:
   - `default` methods (provide a default method body).
   - `static` methods (utility methods belonging to the interface).
5. Cannot have constructors or instance fields.


---
## 1. Basic Interface and Multiple Inheritance
A class can implement multiple interfaces separated by commas.


```java
interface Printable {
    // public static final by default
    int MIN_COPIES = 1;

    // public abstract by default
    void print();
}

interface Scannable {
    void scan();
}

// Implementing multiple interfaces
class AllInOnePrinter implements Printable, Scannable {
    @Override
    public void print() {
        System.out.println("Printing document... (Min copies: " + MIN_COPIES + ")");
    }

    @Override
    public void scan() {
        System.out.println("Scanning document into PDF...");
    }
}

AllInOnePrinter device = new AllInOnePrinter();
device.print();
device.scan();
```

## 2. Default and Static Methods in Interfaces (Java 8+)
- `default` method: Has a body and can be called on an implementing instance or overridden.
- `static` method: Belongs to the interface itself, called as `InterfaceName.methodName()`.


```java
interface Vehicle {
    void start(); // abstract

    // Default method (backward-compatible method with implementation)
    default void horn() {
        System.out.println("Honking horn: Beep Beep!");
    }

    // Static utility method
    static void serviceNotice() {
        System.out.println("Vehicle Service Notice: Check oil every 5,000 km.");
    }
}

class Car implements Vehicle {
    @Override
    public void start() {
        System.out.println("Car engine started with keyless push button.");
    }
}

Car car = new Car();
car.start(); // abstract implementation
car.horn();  // inherited default method
Vehicle.serviceNotice(); // called directly on Interface
```

---
## Summary: Abstract Class vs Interface

| Feature | Abstract Class | Interface |
| :--- | :--- | :--- |
| **Keyword** | `abstract class` | `interface` |
| **Inheritance** | Single inheritance (`extends`) | Multiple inheritance (`implements`) |
| **Methods** | Abstract and non-abstract methods | Mostly abstract (plus `default`/`static` in Java 8+) |
| **Variables** | Any modifier (instance, static, final) | Only `public static final` constants |
| **Constructor** | Yes, has constructors | No constructors |
| **Speed** | Slightly faster than interfaces | Requires search in method table |

## 3. Modern Interface Features: Private Methods (Java 9 & Java 11 - Topic 57)
Java 8 allowed interfaces to have implementation code via `default` and `static` methods. However, if multiple default methods needed common helper code, it could only be exposed publicly.

**Java 9+ introduced `private` and `private static` methods in interfaces**:
- **`private` method**: Encapsulates common reusable logic between multiple `default` methods.
- **`private static` method**: Encapsulates common reusable logic between multiple `static` methods (or default methods).
- Cannot be overridden or accessed by implementing classes or outside callers (pure internal interface encapsulation).


```java
// Demonstrating Java 9+ Private Interface Methods
interface CloudStorageService {
    // Abstract method
    void upload(String fileName);

    // Public default method 1
    default void uploadEncrypted(String fileName) {
        logAction("Starting encrypted upload for: " + fileName);
        // encryption logic
        upload(fileName + ".enc");
    }

    // Public default method 2
    default void uploadCompressed(String fileName) {
        logAction("Starting compressed upload for: " + fileName);
        // compression logic
        upload(fileName + ".zip");
    }

    // Private instance helper method (Java 9+) - shared between default methods
    private void logAction(String action) {
        System.out.println("[CloudStorage Audit Log] " + action + " at timestamp " + System.currentTimeMillis());
    }
}

class AWSStorageServiceImpl implements CloudStorageService {
    @Override
    public void upload(String fileName) {
        System.out.println("File uploaded to AWS S3 bucket: " + fileName);
    }
}

CloudStorageService storage = new AWSStorageServiceImpl();
storage.uploadEncrypted("financial_report.pdf");
storage.uploadCompressed("database_dump.sql");
```
