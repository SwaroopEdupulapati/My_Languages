# Inheritance in Java
**Inheritance** is an OOP mechanism where one class (child / subclass) acquires the properties and methods of another class (parent / superclass) using the `extends` or `implements` keyword.

### Types of Inheritance:
1. **Single Inheritance** (Supported via Classes)
2. **Multilevel Inheritance** (Supported via Classes)
3. **Hierarchical Inheritance** (Supported via Classes)
4. **Multiple Inheritance** (Supported via Interfaces)
5. **Hybrid Inheritance** (Supported via Interfaces)


---
## 0. Introduction to Object-Oriented Programming (OOP) & Principles (Topics 14 & 16)

### A. Procedural Programming (POP) vs. Object-Oriented Programming (OOP)
| Feature | Procedural Programming (e.g. C) | Object-Oriented Programming (Java) |
| :--- | :--- | :--- |
| **Focus** | Focuses on functions and procedures step-by-step | Focuses on real-world entities represented as **Objects** |
| **Data Security** | Data moves freely between functions (low security) | Data is bound tightly to methods via encapsulation (high security) |
| **Approach** | Top-down development approach | Bottom-up modular design approach |
| **Code Reuse** | Limited (relies on function calls) | High (leverages Inheritance and Polymorphism) |

### B. The 4 Core Principles of OOP (Pillars of Java)
1. **Encapsulation**: Wrapping data (variables) and code (methods) together into a single unit (class) and restricting direct access using `private` fields and public getters/setters.
2. **Abstraction**: Hiding internal implementation complexities and exposing only essential functional interfaces to the user (via `abstract` classes and `interface`).
3. **Inheritance**: Mechanism where a child class acquires the fields and methods of a parent class (IS-A relationship), enabling massive code reusability.
4. **Polymorphism**: The ability of an entity to take multiple forms—either at compile-time (Method Overloading) or at runtime (Dynamic Method Overriding).

## 1. Single Inheritance
One child class inherits directly from one parent class.

```
   +------------------+
   |  Animal (Parent) |
   +------------------+
            |
            v (extends)
   +------------------+
   |   Dog (Child)    |
   +------------------+
```


````java
// Parent class
class Animal {
    void eat() {
        System.out.println("Animal is eating.");
    }
}

// Child class
class Dog extends Animal {
    void bark() {
        System.out.println("Dog is barking.");
    }
}

Dog d = new Dog();
d.eat();  // Inherited from Animal
d.bark(); // Defined in Dog
````

---
## 2. Multilevel Inheritance
A class inherits from a derived class, forming an inheritance chain.

```
   +------------------+
   |  Animal (Grand)  |
   +------------------+
            |
            v (extends)
   +------------------+
   |   Dog (Parent)   |
   +------------------+
            |
            v (extends)
   +------------------+
   |  Puppy (Child)   |
   +------------------+
```


````java
class AnimalLevel {
    void eat() {
        System.out.println("Eating food...");
    }
}

class DogLevel extends AnimalLevel {
    void bark() {
        System.out.println("Barking...");
    }
}

class Puppy extends DogLevel {
    void weep() {
        System.out.println("Weeping...");
    }
}

Puppy p = new Puppy();
p.eat();  // From AnimalLevel
p.bark(); // From DogLevel
p.weep(); // From Puppy
````

## 3. Hierarchical Inheritance
Multiple child classes inherit from a single parent class.

```
              +------------------+
              |  Shape (Parent)  |
              +------------------+
                   /        \
        (extends) /          \ (extends)
                 v            v
     +---------------+    +---------------+
     | Circle (Child)|    |Square (Child) |
     +---------------+    +---------------+
```


````java
class Shape {
    void draw() {
        System.out.println("Drawing a shape.");
    }
}

class Circle extends Shape {
    void areaCircle() {
        System.out.println("Calculating circle area.");
    }
}

class Square extends Shape {
    void areaSquare() {
        System.out.println("Calculating square area.");
    }
}

Circle c = new Circle();
Square s = new Square();

c.draw();
c.areaCircle();

s.draw();
s.areaSquare();
````

## 4. Multiple Inheritance (Using Interfaces)
### Note : we will learn this in from interfaces in java it does not support directly by using **extends**.
Java does **not** support multiple inheritance with classes to avoid the Diamond Problem (ambiguity). It is achieved using **interfaces** via the `implements` keyword.

```
   +------------------+     +------------------+
   | Printable (Intf) |     | Showable (Intf)  |
   +------------------+     +------------------+
             \                       /
   (implements\                     / (implements)
               v                   v
             +-----------------------+
             |   Document (Child)    |
             +-----------------------+
```


````java
interface Printable {
    void print();
}

interface Showable {
    void show();
}

// Class implementing multiple interfaces
class Document implements Printable, Showable {
    public void print() {
        System.out.println("Printing document...");
    }

    public void show() {
        System.out.println("Showing document on screen...");
    }
}

Document doc = new Document();
doc.print();
doc.show();
````

## 5. Hybrid Inheritance (Using Classes and Interfaces)
A combination of two or more types of inheritance (e.g. Hierarchical + Multiple). In Java, this is implemented using interfaces.

```
                 +------------------+
                 | Vehicle (Parent) |
                 +------------------+
                          |
                          v (extends)
                 +------------------+
                 |   Car (Class)    |
                 +------------------+
                          |
           +--------------+--------------+
           | (implements)                | (implements)
           v                             v
+--------------------+         +--------------------+
| Electric (Intf)    |         | Autonomous (Intf)  |
+--------------------+         +--------------------+
           \                             /
            \                           /
             v                         v
       +------------------------------------+
       |         TeslaCar (Child)           |
       +------------------------------------+
```


````java
class Vehicle {
    void move() {
        System.out.println("Vehicle is moving.");
    }
}

interface Electric {
    void chargeBattery();
}

interface Autonomous {
    void autoDrive();
}

// Combines class extension with multiple interface implementations
class TeslaCar extends Vehicle implements Electric, Autonomous {
    public void chargeBattery() {
        System.out.println("Battery charging...");
    }

    public void autoDrive() {
        System.out.println("Self-driving mode enabled.");
    }
}

TeslaCar tesla = new TeslaCar();
tesla.move();          // from Vehicle
tesla.chargeBattery(); // from Electric
tesla.autoDrive();     // from Autonomous
````

### 5.1 Hybrid Inheritance Using Only Classes (Hierarchical + Multilevel)
Hybrid inheritance can also be achieved with pure classes by combining **Hierarchical** and **Multilevel** inheritance (avoiding multiple inheritance / diamond problem).

```
                   +-------------------+
                   |  Person (Parent)  |
                   +-------------------+
                             |
            +----------------+----------------+
            | (extends)                       | (extends)
            v                                 v
   +------------------+             +-------------------+
   | Student (Child)  |             | Employee (Child)  |
   +------------------+             +-------------------+
            |                                 |
            v (extends)                       v (extends)
   +------------------+             +-------------------+
   | GraduateStudent  |             |  Manager (Child)  |
   +------------------+             +-------------------+
```

- `Person` &rarr; `Student` & `Employee` forms **Hierarchical Inheritance**.
- `Person` &rarr; `Student` &rarr; `GraduateStudent` forms **Multilevel Inheritance**.
- Combined together, this forms a valid **Hybrid Inheritance** tree without interfaces.


````java
// Base class
class Person {
    void displayPerson() {
        System.out.println("I am a Person.");
    }
}

// Child 1 (Hierarchical branch A)
class Student extends Person {
    void displayStudent() {
        System.out.println("I am a Student.");
    }
}

// Child of Child 1 (Multilevel branch A)
class GraduateStudent extends Student {
    void displayGrad() {
        System.out.println("I am a Graduate Research Student.");
    }
}

// Child 2 (Hierarchical branch B)
class Employee extends Person {
    void displayEmployee() {
        System.out.println("I am an Employee.");
    }
}

// Test Hybrid with Classes Only
GraduateStudent grad = new GraduateStudent();
grad.displayPerson();  // From Grandparent (Person)
grad.displayStudent(); // From Parent (Student)
grad.displayGrad();    // From GraduateStudent

Employee emp = new Employee();
emp.displayPerson();   // From Parent (Person)
emp.displayEmployee(); // From Employee
````

## 6. The `super` Keyword
The `super` keyword in Java is a reference variable used to refer directly to the immediate parent (superclass) object.

### 3 Main Uses of `super`:
1. **`super.variable`**: Access immediate parent class fields when shadowed by child class.
2. **`super.method()`**: Invoke immediate parent class methods when overridden in child class.
3. **`super()`**: Call immediate parent class constructor (must be the first statement in child constructor).


````java
// Parent Class
class VehicleParent {
    String brand = "Generic Brand";

    VehicleParent(String brand) {
        this.brand = brand;
        System.out.println("Parent Constructor: Vehicle brand set to " + this.brand);
    }

    void displaySpeed() {
        System.out.println("Parent: Vehicle max speed is 120 km/h.");
    }
}

// Child Class using super
class CarChild extends VehicleParent {
    String brand = "Sports Edition"; // shadows parent field

    CarChild(String parentBrand, String childBrand) {
        // 1. super() - calls parent constructor (must be first line)
        super(parentBrand);
        this.brand = childBrand;
    }

    @Override
    void displaySpeed() {
        System.out.println("Child: Car max speed is 240 km/h.");
    }

    void showDemo() {
        // 2. super.variable vs this.variable
        System.out.println("this.brand  : " + this.brand);
        System.out.println("super.brand : " + super.brand);

        // 3. super.method() vs this.method()
        this.displaySpeed();  // calls overridden child method
        super.displaySpeed(); // calls parent method
    }
}

// Testing 'super'
CarChild car = new CarChild("Toyota", "Supra");
car.showDemo();
````

## 7. Association, Aggregation, and Composition (HAS-A Relationships)
While Inheritance represents an **IS-A** relationship, Object Design also relies heavily on **HAS-A** relationships:

1. **Association**: General connection/relationship between two separate classes (e.g., `Teacher` teaches `Student`).
2. **Aggregation (Weak HAS-A)**: A class contains references to other independent objects. If the container is destroyed, the contained objects can still exist independently (e.g., `Department` has `Teachers`; if `Department` closes, `Teachers` still exist).
3. **Composition (Strong HAS-A)**: The contained object cannot exist independently of its parent. If the parent is destroyed, the child objects are destroyed too (e.g., `Car` has an `Engine`; if `Car` is scrapped, its `Engine` lifecycle ends with it).


````java
// 1. Aggregation (Weak HAS-A): Teacher exists independently of Department
class Teacher {
    String teacherName;
    Teacher(String name) { this.teacherName = name; }
}

class Department {
    String deptName;
    Teacher headTeacher; // Aggregation
    Department(String deptName, Teacher teacher) {
        this.deptName = deptName;
        this.headTeacher = teacher;
    }
}

// 2. Composition (Strong HAS-A): Engine created and owned entirely by Car
class Engine {
    String type;
    Engine(String type) { this.type = type; }
}

class Automobile {
    private final Engine engine; // Composition: tightly bound lifecycle
    Automobile() {
        this.engine = new Engine("V8 Turbo");
    }
    void showDetails() {
        System.out.println("Automobile running with: " + engine.type);
    }
}

// Demonstration
Teacher t = new Teacher("Dr. Sharma");
Department d = new Department("Computer Science", t);
System.out.println("Aggregation: " + d.deptName + " headed by " + d.headTeacher.teacherName);

Automobile car = new Automobile();
car.showDetails();
````

## 8. Access Modifiers Summary Table
Java provides 4 access levels to enforce encapsulation and modularity:

| Modifier | Inside Same Class | Inside Same Package | Outside Package by Subclass | Outside Package Anywhere |
| :--- | :---: | :---: | :---: | :---: |
| **`private`** | ✅ Yes | ❌ No | ❌ No | ❌ No |
| **`default` (package-private)** | ✅ Yes | ✅ Yes | ❌ No | ❌ No |
| **`protected`** | ✅ Yes | ✅ Yes | ✅ Yes (via inheritance) | ❌ No |
| **`public`** | ✅ Yes | ✅ Yes | ✅ Yes | ✅ Yes |

## 9. Accessing Protected Variables and Methods Outside the Package (Topic 41)
The `protected` access modifier is specifically designed for inheritance across package boundaries:

1. **Inside Same Package**: A `protected` member behaves like `default`—it is accessible by any class in the same package.
2. **Outside the Package**: A `protected` member is accessible **ONLY to derived subclasses** through inheritance (`extends`).
3. **Critical Rule for Outside Package Access**:
   - The child class can access inherited `protected` members via `this`, `super`, or a child class reference (`childObj.protectedMember`).
   - **Direct Parent Reference Restriction**: A subclass **CANNOT** access `protected` members using a parent class instance (`parentObj.protectedMember` will fail compilation outside the package)!


````java
// Simulating Package A: com.cdac.security
class SecurityVaultParent {
    // Protected variable and method
    protected String securityKey = "SEC-KEY-99482";

    protected void displayVaultSecurity() {
        System.out.println("Vault Security Key: " + securityKey);
    }
}

// Simulating Package B: com.cdac.client (Subclass in another package)
class AuthorizedClientChild extends SecurityVaultParent {
    public void accessProtectedMembers() {
        // 1. Legal: Accessing inherited protected member directly or via super
        System.out.println("Access via Subclass Inheritance: " + this.securityKey);
        super.displayVaultSecurity(); // Legal

        // 2. Note on Illegal Cross-Package Access:
        // SecurityVaultParent p = new SecurityVaultParent();
        // p.securityKey; // COMPILATION ERROR outside package: securityKey has protected access!
    }
}

AuthorizedClientChild client = new AuthorizedClientChild();
client.accessProtectedMembers();
````
