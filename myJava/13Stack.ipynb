# Stack in Java (`java.util.Stack`)
A **Stack** represents a **Last-In, First-Out (LIFO)** data structure. Because `Stack` extends `Vector`, it inherits both stack-specific methods and all list/vector methods.

### Complete Methods Reference:
| Stack-Specific Method | Description |
| :--- | :--- |
| `push(item)` | Pushes item onto the top of the stack |
| `pop()` | Removes and returns the top item |
| `peek()` | Inspects top item without removing |
| `empty()` / `isEmpty()` | Checks whether the stack has 0 elements |
| `search(Object)` | Returns 1-based distance from top (-1 if absent) |

| Inherited Vector/Collection Methods | Description |
| :--- | :--- |
| `size()` | Total count of elements |
| `contains(Object)` | Check if element exists in stack |
| `indexOf(Object)` / `elementAt(index)` | Access elements by 0-based position |
| `firstElement()` / `lastElement()` | Access bottom (`first`) or top (`last`) element |
| `clear()` | Remove all items |


```java
import java.util.Stack;

Stack<Integer> st = new Stack<>();

// 1. Core Stack Operations
st.push(10);
st.push(20);
st.push(30);
st.push(40);
System.out.println("Stack: " + st);

// peek()
System.out.println("peek() (Top Element): " + st.peek());

// pop()
System.out.println("pop()  (Removed Top) : " + st.pop());
System.out.println("Stack after pop: " + st);

// search() - 1-based index from top of stack
System.out.println("search(30) [Top is 1]: " + st.search(30)); // 1
System.out.println("search(10) [Bottom]   : " + st.search(10)); // 3
System.out.println("search(99) [Not found]: " + st.search(99)); // -1

// 2. Inherited Vector methods
System.out.println("size(): " + st.size());
System.out.println("empty(): " + st.empty() + ", isEmpty(): " + st.isEmpty());
System.out.println("firstElement() (Bottom): " + st.firstElement());
System.out.println("lastElement()  (Top)   : " + st.lastElement());
System.out.println("elementAt(1): " + st.elementAt(1));
System.out.println("contains(20): " + st.contains(20));

// 3. Clear
st.clear();
System.out.println("After clear(), empty(): " + st.empty());
```
