# Encapsulation in Java
**Encapsulation** is wrapping data (fields) and methods operating on that data into a single unit (class), while hiding direct access to the fields (data hiding).

### How to Achieve Encapsulation:
1. Declare variables as `private` (cannot be accessed directly from outside the class).
2. Provide public `getter` and `setter` methods to read and modify field values with validation.


---
## Encapsulated Class Example
A `BankAccount` class where `balance` is hidden and can only be modified through controlled rules.


```java
class BankAccount {
    // 1. Private fields (data hiding)
    private String accountHolder;
    private double balance;

    // Constructor
    public BankAccount(String accountHolder, double initialBalance) {
        this.accountHolder = accountHolder;
        if (initialBalance >= 0) {
            this.balance = initialBalance;
        } else {
            this.balance = 0.0;
        }
    }

    // 2. Getter methods (read-only access)
    public String getAccountHolder() {
        return this.accountHolder;
    }

    public double getBalance() {
        return this.balance;
    }

    // 3. Setter methods with validation rules
    public void setAccountHolder(String accountHolder) {
        if (accountHolder != null && !accountHolder.trim().isEmpty()) {
            this.accountHolder = accountHolder;
        } else {
            System.out.println("Invalid name provided!");
        }
    }

    public void deposit(double amount) {
        if (amount > 0) {
            this.balance += amount;
            System.out.println("Deposited: Rs. " + amount);
        } else {
            System.out.println("Deposit amount must be positive.");
        }
    }
}

// Testing Encapsulation
BankAccount account = new BankAccount("Swaroop", 5000.0);

// Reading data through getters
System.out.println("Account Holder : " + account.getAccountHolder());
System.out.println("Current Balance: Rs. " + account.getBalance());

// Modifying data safely through controlled methods
account.deposit(2500.0);
System.out.println("Updated Balance: Rs. " + account.getBalance());

// Direct field access is blocked (compile-time error if uncommented):
// account.balance = -99999; // Error: balance has private access in BankAccount
```

## Read-Only and Write-Only Classes
- **Read-Only Class**: Provide only `getter` methods (no setters).
- **Write-Only Class**: Provide only `setter` methods (no getters).


```java
// Read-only class (Immutable view)
class StudentRecord {
    private int id = 101;
    private String college = "CDAC";

    // Only getters, values cannot be changed from outside
    public int getId() { return id; }
    public String getCollege() { return college; }
}

StudentRecord record = new StudentRecord();
System.out.println("ID: " + record.getId() + ", College: " + record.getCollege());
```
