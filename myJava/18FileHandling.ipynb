# File Handling in Java
File handling lets us create, read, and write files step by step.

### Covered Topics:
1. **File Info**: Check if a file exists, get name and size.
2. **Text Files**: Write and read simple lines using `FileWriter` and `FileReader`.
3. **Buffered Text**: Fast writing and reading using `BufferedWriter` and `BufferedReader`.
4. **Binary Files**: Write and read raw bytes using `FileOutputStream` and `FileInputStream`.
5. **Object Files**: Save and load objects using `ObjectOutputStream` and `ObjectInputStream`.
6. **CSV Files**: Save and read comma-separated data in clear steps.


---
## 1. File Information (`File` class)
Create a file reference and inspect its details.


```java
import java.io.File;
import java.io.IOException;

// Step 1: Create File object with file name
String path = "sample.txt";
File myFile = new File(path);

try {
    // Step 2: Create the file on disk
    boolean isCreated = myFile.createNewFile();
    System.out.println("File created: " + isCreated);
    
    // Step 3: Check details
    System.out.println("Name     : " + myFile.getName());
    System.out.println("Exists   : " + myFile.exists());
    System.out.println("Size     : " + myFile.length() + " bytes");
    System.out.println("Can Read : " + myFile.canRead());
    System.out.println("Can Write: " + myFile.canWrite());
} catch (IOException e) {
    System.out.println("Error: " + e.getMessage());
}
```

## 2. Text Files: `FileWriter` and `FileReader`
Write text and read it character by character in simple steps.


```java
import java.io.File;
import java.io.FileWriter;
import java.io.FileReader;
import java.io.IOException;

// Step 1: Target file
File textFile = new File("hello.txt");

// Step 2: Open FileWriter and write short words
try {
    FileWriter fw = new FileWriter(textFile);
    
    fw.write("Java\n");
    fw.write("Python\n");
    fw.write("C++\n");
    
    fw.close(); // Save and close
    System.out.println("Text written to file.");
} catch (IOException e) {
    System.out.println("Write error: " + e.getMessage());
}

// Step 3: Open FileReader and read text back
try {
    FileReader fr = new FileReader(textFile);
    
    int ch;
    System.out.print("Reading file: ");
    while ((ch = fr.read()) != -1) {
        System.out.print((char) ch);
    }
    
    fr.close(); // Close reader
} catch (IOException e) {
    System.out.println("Read error: " + e.getMessage());
}
```

## 3. Buffered Text: `BufferedWriter` and `BufferedReader`
Using buffers is faster and allows writing and reading line by line.


```java
import java.io.File;
import java.io.FileWriter;
import java.io.BufferedWriter;
import java.io.FileReader;
import java.io.BufferedReader;
import java.io.IOException;

// Step 1: Target file
File notesFile = new File("notes.txt");

// Step 2: Create FileWriter, then wrap in BufferedWriter
try {
    FileWriter fw = new FileWriter(notesFile);
    BufferedWriter bw = new BufferedWriter(fw);
    
    bw.write("First line");
    bw.newLine();
    bw.write("Second line");
    bw.newLine();
    bw.write("Third line");
    
    bw.close(); // Closes bw and fw
    System.out.println("Lines written with buffer.");
} catch (IOException e) {
    System.out.println("Write error: " + e.getMessage());
}

// Step 3: Create FileReader, then wrap in BufferedReader
try {
    FileReader fr = new FileReader(notesFile);
    BufferedReader br = new BufferedReader(fr);
    
    String line;
    System.out.println("\n--- Reading Lines ---");
    while ((line = br.readLine()) != null) {
        System.out.println(line);
    }
    
    br.close(); // Closes br and fr
} catch (IOException e) {
    System.out.println("Read error: " + e.getMessage());
}
```

## 4. Binary Data: `FileOutputStream` and `FileInputStream`
Used for raw bytes (like data files, images, or audio).


```java
import java.io.File;
import java.io.FileOutputStream;
import java.io.FileInputStream;
import java.io.IOException;

// Step 1: Target file
File binFile = new File("bytes.bin");

// Step 2: Write raw bytes
try {
    FileOutputStream fos = new FileOutputStream(binFile);
    
    byte[] numbers = {10, 20, 30, 40, 50};
    fos.write(numbers);
    
    fos.close();
    System.out.println("Bytes written successfully.");
} catch (IOException e) {
    System.out.println("Write error: " + e.getMessage());
}

// Step 3: Read raw bytes
try {
    FileInputStream fis = new FileInputStream(binFile);
    
    int b;
    System.out.print("Read bytes: ");
    while ((b = fis.read()) != -1) {
        System.out.print(b + " ");
    }
    System.out.println();
    
    fis.close();
} catch (IOException e) {
    System.out.println("Read error: " + e.getMessage());
}
```

## 5. Serialization and Deserialization (`Serializable` and `transient`)
- **Serialization**: Saving an object to a file with `ObjectOutputStream`.
- **Deserialization**: Loading the object back with `ObjectInputStream`.
- **`transient` Keyword**: Skips fields during serialization so they are not saved.


```java
import java.io.File;
import java.io.Serializable;
import java.io.FileOutputStream;
import java.io.ObjectOutputStream;
import java.io.FileInputStream;
import java.io.ObjectInputStream;

// Step 1: Class implementing Serializable with a transient variable
class User implements Serializable {
    private static final long serialVersionUID = 1L;
    int id;
    String name;
    transient String password; // transient: will NOT be saved to file

    User(int id, String name, String password) {
        this.id = id;
        this.name = name;
        this.password = password;
    }
}

// Step 2: File reference and Object
File objFile = new File("user.dat");
User u1 = new User(101, "Swaroop", "mySecret123");

// Step 3: Write object to file (Serialization)
try {
    FileOutputStream fos = new FileOutputStream(objFile);
    ObjectOutputStream oos = new ObjectOutputStream(fos);
    
    oos.writeObject(u1);
    
    oos.close();
    fos.close();
    System.out.println("User object serialized to file.");
} catch (Exception e) {
    System.out.println("Save error: " + e.getMessage());
}

// Step 4: Read object from file (Deserialization)
try {
    FileInputStream fis = new FileInputStream(objFile);
    ObjectInputStream ois = new ObjectInputStream(fis);
    
    User loadedUser = (User) ois.readObject();
    System.out.println("Loaded User ID   : " + loadedUser.id);
    System.out.println("Loaded User Name : " + loadedUser.name);
    System.out.println("Password (transient): " + loadedUser.password); // prints null
    
    ois.close();
    fis.close();
} catch (Exception e) {
    System.out.println("Load error: " + e.getMessage());
}
```

## 6. CSV Files: Comma-Separated Values
Clear, separated steps to create, write, and parse a `.csv` file.


```java
import java.io.File;
import java.io.FileWriter;
import java.io.FileReader;
import java.io.BufferedReader;
import java.io.IOException;

// Step 1: Create File reference
File csvFile = new File("students.csv");

// Step 2: Open FileWriter separately and append lines
try {
    FileWriter fw = new FileWriter(csvFile);
    
    // Write header
    fw.append("ID,Name,Marks\n");
    
    // Write rows with simple short words
    fw.append("1,Alex,85\n");
    fw.append("2,Bob,90\n");
    fw.append("3,John,78\n");
    
    fw.close(); // Save and close
    System.out.println("CSV file created.");
} catch (IOException e) {
    System.out.println("Write error: " + e.getMessage());
}

// Step 3: Open FileReader and BufferedReader in separate lines
try {
    FileReader fr = new FileReader(csvFile);
    BufferedReader br = new BufferedReader(fr);
    
    String row;
    System.out.println("\n--- Reading CSV Rows ---");
    while ((row = br.readLine()) != null) {
        String[] items = row.split(",");
        System.out.println("ID: " + items[0] + " | Name: " + items[1] + " | Marks: " + items[2]);
    }
    
    br.close();
    fr.close();
} catch (IOException e) {
    System.out.println("Read error: " + e.getMessage());
}

// Step 4: Cleanup demo files
csvFile.delete();
new File("sample.txt").delete();
new File("hello.txt").delete();
new File("notes.txt").delete();
new File("bytes.bin").delete();
new File("user.dat").delete();
System.out.println("\nTemporary files cleaned up.");
```

## 7. Modern Java NIO Package Overview (`java.nio.file`) (Topic 72)
**Java NIO (New I/O / Non-blocking I/O)** was introduced to overcome limitations of traditional stream-based `java.io`:

### Key Architectural Differences:
| Feature | Traditional `java.io` | Modern `java.nio` (NIO.2) |
| :--- | :--- | :--- |
| **Mechanism** | Stream-oriented (reads byte-by-byte or char-by-char) | **Buffer and Channel oriented** (reads high-speed blocks) |
| **Blocking** | Strict blocking I/O (threads wait for data) | Non-blocking / Asynchronous I/O supported |
| **Path Handling**| Legacy `java.io.File` class | **`java.nio.file.Path`** and **`Paths`** |
| **File Operations**| Complex multi-line boilerplate streams | High-level atomic static methods in **`java.nio.file.Files`** |

### Core NIO Components:
1. **`Path`**: Represents hierarchical file/directory paths.
2. **`Files`**: Rich utility class with static methods for reading, writing, copying, moving, and streaming lines.
3. **`ByteBuffer`**: Direct memory buffer for fast low-level data transfers.
4. **`Channel`**: Two-way open connection to a hardware device, file, or socket.


```java
import java.nio.file.Path;
import java.nio.file.Paths;
import java.nio.file.Files;
import java.nio.file.StandardOpenOption;
import java.nio.ByteBuffer;
import java.io.IOException;
import java.util.List;

try {
    // 1. Path creation
    Path nioFile = Paths.get("nio_demo_output.txt");
    System.out.println("Path resolved: " + nioFile.toAbsolutePath());

    // 2. Writing String to file in 1 single line (Java 11+ Files.writeString)
    String content = "Java NIO makes file handling clean, fast, and robust!\nSecond Line from NIO.";
    Files.writeString(nioFile, content, StandardOpenOption.CREATE, StandardOpenOption.WRITE, StandardOpenOption.TRUNCATE_EXISTING);
    System.out.println("File written successfully using Files.writeString!");

    // 3. Reading String in 1 line
    String readBack = Files.readString(nioFile);
    System.out.println("\n--- Read Content via Files.readString() ---\n" + readBack);

    // 4. Reading all lines into a List or Stream
    List<String> lines = Files.readAllLines(nioFile);
    System.out.println("\nTotal Lines read via Files.readAllLines(): " + lines.size());

    // 5. NIO ByteBuffer Demonstration (Allocate, Put, Flip, Get)
    ByteBuffer buffer = ByteBuffer.allocate(16); // 16 bytes buffer
    buffer.put((byte) 'N');
    buffer.put((byte) 'I');
    buffer.put((byte) 'O');
    buffer.flip(); // switch from writing mode to reading mode
    System.out.print("ByteBuffer bytes: ");
    while (buffer.hasRemaining()) {
        System.out.print((char) buffer.get() + " ");
    }
    System.out.println();

    // Cleanup demo file
    Files.deleteIfExists(nioFile);
} catch (IOException e) {
    System.out.println("NIO Exception: " + e.getMessage());
}
```
