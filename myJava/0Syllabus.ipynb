## 🎯 CDAC Evaluation & Technical Interview Focus Blueprint

Key theoretical and practical evaluation themes prioritized in assessments:

1. **Object-Oriented Core & Memory Mechanics**:
   - Stack vs Heap memory allocation during object instantiation
   - Constructor chaining sequence with `super()` and `this()`
   - Method Overloading (compile-time binding) vs Overriding (runtime virtual method invocation)
   - Abstract Classes vs Interfaces (especially Java 8/11 enhancements: default, static, private methods)

2. **The Collections Framework & Hashing Deep Dive**:
   - Deep understanding of `equals()` and `hashCode()` contract (why hash codes must match for equal objects)
   - HashMap internal bucket layout (Array of LinkedList / Red-Black Tree in Java 8+)
   - Difference between fail-fast (`ArrayList`, `HashMap`) and fail-safe/thread-safe iterators (`ConcurrentHashMap`)
   - Natural sorting with `Comparable.compareTo()` vs customizable sorting with `Comparator.compare()`

3. **Modern Functional Java & Streams (Java 8 to 21)**:
   - The four core functional interfaces: `Predicate<T>`, `Function<T,R>`, `Consumer<T>`, `Supplier<T>`
   - Stream lazy evaluation: Intermediate (stateless/stateful) vs Terminal operations
   - Reductions and advanced grouping using `Collectors.groupingBy()` and `partitioningBy()`
   - Modern syntax: Pattern Matching for `instanceof`, Sealed Classes, and Records

4. **Concurrency, Exceptions & Persistence**:
   - Exception handling propagation semantics and Checked vs Unchecked recovery rules
   - Thread synchronization, intrinsic locks (monitor), and thread communication (`wait`/`notify`)
   - Secure database access via JDBC `PreparedStatement` to prevent SQL Injection vulnerabilities
   - The Data Access Object (DAO) pattern isolating business logic from persistence
