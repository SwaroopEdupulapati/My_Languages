# HashMap in Java (`java.util.HashMap`)
A **HashMap** stores data in **Key-Value pairs** (`<K, V>`). It implements the `Map` interface and uses a hashing technique to store keys.

### Key Characteristics:
- **Unique Keys**: Keys must be unique; duplicate keys overwrite the existing value.
- **Values**: Can be duplicated.
- **Nulls**: Allows **one `null` key** and **multiple `null` values**.
- **Unordered**: Does not guarantee any order of keys or values.
- **Performance**: $O(1)$ average time complexity for `get()` and `put()` operations.

### Complete Methods Reference:
| Method | Description |
| :--- | :--- |
| `put(key, value)` | Associates value with key (inserts or updates) |
| `putIfAbsent(key, value)` | Inserts only if key is not already present or is null |
| `putAll(map)` | Copies all mappings from another map |
| `get(key)` | Returns value mapped to key, or `null` if absent |
| `getOrDefault(key, defaultVal)` | Returns value or specified default if absent |
| `containsKey(key)` | Checks if the key exists |
| `containsValue(value)` | Checks if the value exists |
| `remove(key)` | Removes mapping for a key |
| `remove(key, value)` | Removes only if key is currently mapped to specified value |
| `replace(key, val)` / `replace(k, oldV, newV)` | Replaces existing mapping |
| `keySet()` | Returns a `Set` view of all keys |
| `values()` | Returns a `Collection` view of all values |
| `entrySet()` | Returns a `Set` view of `Map.Entry<K, V>` mappings |
| `size()` / `isEmpty()` | Returns pair count / checks empty state |
| `clear()` | Removes all mappings |
| `computeIfAbsent()` / `merge()` | Lambda-based atomic computations |


```java
import java.util.HashMap;
import java.util.Map;

// 1. Creation and put() operations
HashMap<String, Integer> map = new HashMap<>();
map.put("Java", 1995);
map.put("Python", 1991);
map.put("C++", 1985);
map.put("Java", 1996); // overwrites previous 'Java' value
map.put(null, 0);       // one null key allowed
System.out.println("HashMap: " + map);

// 2. putIfAbsent()
map.putIfAbsent("Python", 2000); // won't overwrite
map.putIfAbsent("Rust", 2010);   // will insert
System.out.println("After putIfAbsent: " + map);

// 3. Retrieval methods
System.out.println("get('Java'): " + map.get("Java"));
System.out.println("getOrDefault('Go', 2009): " + map.getOrDefault("Go", 2009));

// 4. Inspection
System.out.println("containsKey('C++'): " + map.containsKey("C++"));
System.out.println("containsValue(1991): " + map.containsValue(1991));
System.out.println("size(): " + map.size() + ", isEmpty(): " + map.isEmpty());

// 5. Update and Replace
map.replace("Rust", 2015);
map.replace("C++", 1985, 1983); // conditional replace
System.out.println("After replace: " + map);

// 6. Removal
map.remove(null);
map.remove("C++", 1999); // fails: value doesn't match
map.remove("C++", 1983); // succeeds
System.out.println("After removals: " + map);

// 7. Iteration via keySet(), values(), and entrySet()
System.out.println("\n--- Iteration via entrySet() ---");
for (Map.Entry<String, Integer> entry : map.entrySet()) {
    System.out.println(entry.getKey() + " -> " + entry.getValue());
}

// 8. Clear
map.clear();
System.out.println("After clear(), isEmpty(): " + map.isEmpty());
```

## 2. `LinkedHashMap`, `TreeMap`, and `Hashtable` (The Map Hierarchy)

| Feature | `HashMap` | `LinkedHashMap` | `TreeMap` | `Hashtable` |
| :--- | :--- | :--- | :--- | :--- |
| **Order** | Unordered | **Maintains Insertion Order** | **Sorted by Keys** | Unordered |
| **Thread-Safe** | ❌ No | ❌ No | ❌ No | ✅ Yes (Synchronized) |
| **Null Key/Value**| 1 null key, many null values | 1 null key, many null values | ❌ No null keys | ❌ No null key or value |
| **Backing Structure** | Hash Table | Hash Table + Linked List | Red-Black Tree | Hash Table |
| **Performance** | $O(1)$ | $O(1)$ | $O(\log n)$ | $O(1)$ (synchronized overhead) |


```java
import java.util.HashMap;
import java.util.LinkedHashMap;
import java.util.TreeMap;
import java.util.Hashtable;

// 1. LinkedHashMap (Preserves insertion order)
LinkedHashMap<Integer, String> linkedMap = new LinkedHashMap<>();
linkedMap.put(103, "Charlie");
linkedMap.put(101, "Alice");
linkedMap.put(102, "Bob");
System.out.println("LinkedHashMap (Preserves Order): " + linkedMap);

// 2. TreeMap (Sorts keys ascending)
TreeMap<Integer, String> treeMap = new TreeMap<>();
treeMap.put(103, "Charlie");
treeMap.put(101, "Alice");
treeMap.put(102, "Bob");
System.out.println("TreeMap (Keys Sorted Ascending): " + treeMap);
System.out.println("First Key: " + treeMap.firstKey() + ", Last Key: " + treeMap.lastKey());

// 3. Hashtable (Legacy thread-safe map)
Hashtable<String, Integer> hashtable = new Hashtable<>();
hashtable.put("A", 10);
hashtable.put("B", 20);
System.out.println("Hashtable: " + hashtable);
```
