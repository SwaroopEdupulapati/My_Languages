# ArrayList in Java (`java.util.ArrayList`)
An **ArrayList** is a resizable, dynamic array implementation of the `List` interface.

### Complete Methods Reference:
| Method | Description |
| :--- | :--- |
| `add(e)` / `add(index, e)` | Append element / insert at index |
| `addAll(collection)` / `addAll(index, c)` | Append / insert another collection |
| `get(index)` | Access element by index |
| `set(index, e)` | Replace/update element at index |
| `remove(index)` / `remove(Object)` | Remove by index or by element matching |
| `removeIf(predicate)` | Remove elements matching a condition |
| `indexOf(Object)` / `lastIndexOf(Object)` | First / last index of an element |
| `contains(Object)` / `containsAll(c)` | Check presence of element / sub-collection |
| `size()` / `isEmpty()` | Number of elements / check empty state |
| `subList(from, to)` | Extract view of range `[from, to)` |
| `toArray()` | Convert to array (`Object[]` or typed) |
| `clone()` | Creates shallow copy of the list |
| `ensureCapacity(min)` / `trimToSize()` | Memory & internal buffer tuning |
| `clear()` | Remove all elements |
| `sort(comparator)` | In-place sort using Comparator |


```java
import java.util.ArrayList;
import java.util.Arrays;
import java.util.Collections;

// 1. Initialization and adding elements
ArrayList<String> list = new ArrayList<>();
list.add("Java");
list.add("Python");
list.add("C++");
list.add(1, "Kotlin"); // insert at index 1

ArrayList<String> extra = new ArrayList<>(Arrays.asList("Go", "Rust"));
list.addAll(extra); // add entire collection
System.out.println("List after addAll: " + list);

// 2. Access and Inspection
System.out.println("get(2): " + list.get(2));
System.out.println("indexOf('Kotlin'): " + list.indexOf("Kotlin"));
System.out.println("contains('Rust'): " + list.contains("Rust"));
System.out.println("size(): " + list.size() + ", isEmpty(): " + list.isEmpty());

// 3. Modification and SubList
list.set(0, "Modern Java"); // update index 0
System.out.println("subList(1, 4): " + list.subList(1, 4));

// 4. Removal operations
list.remove("C++");        // remove by object
list.remove(2);            // remove by index
list.removeIf(s -> s.startsWith("P")); // conditional removal
System.out.println("After removals: " + list);

// 5. toArray and clone
String[] strArr = list.toArray(new String[0]);
System.out.println("Converted to Array length: " + strArr.length);
@SuppressWarnings("unchecked")
ArrayList<String> clonedList = (ArrayList<String>) list.clone();
System.out.println("Cloned list: " + clonedList);

// 6. Sorting and Clearing
Collections.sort(list);
System.out.println("Sorted: " + list);
list.clear();
System.out.println("After clear(): " + list + ", isEmpty(): " + list.isEmpty());
```
