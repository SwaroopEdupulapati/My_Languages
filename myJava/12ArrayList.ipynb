## 0. The `Object` Class in Java (`java.lang.Object`) (Topic 78)

### Complete Summary of the 11 Methods in `java.lang.Object`:

| `notifyAll()` | `void` | Wakes up all threads waiting on this object's monitor |

## 0.1 Collection Framework & Hierarchy Overview (Topics 83, 84 & 85)
The **Java Collection Framework (JCF)** (`java.util`) provides a unified architecture for storing and manipulating groups of objects dynamically without fixed array sizing limitations.

### Complete Collection Framework Hierarchy Diagram (Topic 84):
```
                              Iterable<E> (Interface)
                                   |
                              Collection<E> (Interface)
             /                             |                            \
       List<E>                         Queue<E>                        Set<E>
    /     |    \                     /          \                     /      \
ArrayList LinkedList Vector      PriorityQueue  Deque<E>          HashSet    SortedSet<E>
                       |                          |                  |            |
                     Stack                    ArrayDeque       LinkedHashSet   TreeSet

  ------------------------------------------------------------------------------------
  Map<K, V> (Independent Hierarchy - Key-Value Pairs, does NOT inherit Collection)
    ├── HashMap          (Fastest O(1), unordered, allows 1 null key)
    ├── LinkedHashMap    (Maintains insertion order via doubly-linked list)
    ├── TreeMap          (Implements SortedMap, sorted naturally by key in Red-Black Tree)
    └── Hashtable        (Legacy, thread-safe synchronized methods, no null keys/values)
```

### Comparison of the 4 Core Collection Interfaces (Topic 85):
| Feature | `List` | `Set` | `Queue` | `Map` |
| :--- | :--- | :--- | :--- | :--- |
| **Duplicates** | Allowed | **Not Allowed (Unique)**| Allowed | Keys: Unique, Values: Duplicate |
| **Order** | Preserved (Index-based) | Unordered (except LinkedHashSet/TreeSet)| FIFO / Priority-based | Unordered (except LinkedHashMap/TreeMap) |
| **Null Elements** | Multiple allowed | At most 1 (HashSet/LinkedHashSet)| Disallowed in PriorityQueue | 1 null key (HashMap), multiple null values |
| **Access Mechanism**| By zero-based index | By iterator / contains | `poll()`, `peek()`, `element()`| By Key (`map.get(key)`) |


# ArrayList in Java (`java.util.ArrayList`)
An **ArrayList** is a resizable, dynamic array implementation of the `List` interface.

### Complete Methods Reference:
| Method | Description |
| :--- | :--- |
| `add(e)` / `add(index, e)` | Append element / insert at index |
| `addAll(collection)` / `addAll(index, c)` | Append / insert another collection |
| `get(index)` | Access element by index |
| `set(index, e)` | Replace/update element at index |
| `remove(index)` / `remove(Object)` | Remove by index or by element matching |
| `removeIf(predicate)` | Remove elements matching a condition |
| `indexOf(Object)` / `lastIndexOf(Object)` | First / last index of an element |
| `contains(Object)` / `containsAll(c)` | Check presence of element / sub-collection |
| `size()` / `isEmpty()` | Number of elements / check empty state |
| `subList(from, to)` | Extract view of range `[from, to)` |
| `toArray()` | Convert to array (`Object[]` or typed) |
| `clone()` | Creates shallow copy of the list |
| `ensureCapacity(min)` / `trimToSize()` | Memory & internal buffer tuning |
| `clear()` | Remove all elements |
| `sort(comparator)` | In-place sort using Comparator |


````java
import java.util.ArrayList;
import java.util.Arrays;
import java.util.Collections;

// 1. Initialization and adding elements
ArrayList<String> list = new ArrayList<>();
list.add("Java");
list.add("Python");
list.add("C++");
list.add(1, "Kotlin"); // insert at index 1

ArrayList<String> extra = new ArrayList<>(Arrays.asList("Go", "Rust"));
list.addAll(extra); // add entire collection
System.out.println("List after addAll: " + list);

// 2. Access and Inspection
System.out.println("get(2): " + list.get(2));
System.out.println("indexOf('Kotlin'): " + list.indexOf("Kotlin"));
System.out.println("contains('Rust'): " + list.contains("Rust"));
System.out.println("size(): " + list.size() + ", isEmpty(): " + list.isEmpty());

// 3. Modification and SubList
list.set(0, "Modern Java"); // update index 0
System.out.println("subList(1, 4): " + list.subList(1, 4));

// 4. Removal operations
list.remove("C++");        // remove by object
list.remove(2);            // remove by index
list.removeIf(s -> s.startsWith("P")); // conditional removal
System.out.println("After removals: " + list);

// 5. toArray and clone
String[] strArr = list.toArray(new String[0]);
System.out.println("Converted to Array length: " + strArr.length);
@SuppressWarnings("unchecked")
ArrayList<String> clonedList = (ArrayList<String>) list.clone();
System.out.println("Cloned list: " + clonedList);

// 6. Sorting and Clearing
Collections.sort(list);
System.out.println("Sorted: " + list);
list.clear();
System.out.println("After clear(): " + list + ", isEmpty(): " + list.isEmpty());
````

## 2. Legacy Collection: `Vector`
- **`Vector`**: Similar to `ArrayList`, but its methods are **synchronized** (thread-safe).
- Reallocates by **doubling (100%)** capacity when full (unlike `ArrayList` which grows by 50%).
- In modern Java, `ArrayList` is preferred unless thread-safety is required, where `CopyOnWriteArrayList` or `Collections.synchronizedList()` is chosen.


````java
import java.util.Vector;

Vector<String> vec = new Vector<>(3); // initial capacity 3
vec.add("Delhi");
vec.add("Mumbai");
vec.add("Pune");

System.out.println("Vector size    : " + vec.size());
System.out.println("Vector capacity: " + vec.capacity());

vec.add("Bengaluru"); // triggers capacity doubling
System.out.println("Capacity after growth: " + vec.capacity());
System.out.println("Vector contents      : " + vec);
````

## 2.1 The `java.util.Collections` Utility Class (Topic 87)
`Collections` is a utility class consisting exclusively of static methods that operate on or return collections:
- **`sort(list)`**: Natural or custom Comparator sorting.
- **`reverse(list)`**: Reverses the order of the elements in-place.
- **`shuffle(list)`**: Randomly permutes the specified list using a default source of randomness.
- **`binarySearch(list, key)`**: Fast $O(\log n)$ lookup on sorted lists.
- **`min(col)` / `max(col)`**: Returns the minimum or maximum element.
- **`frequency(col, o)`**: Counts the occurrences of an element in a collection.
- **`disjoint(c1, c2)`**: Returns `true` if the two specified collections have no elements in common.
- **`unmodifiableList(list)`**: Wraps the collection in an immutable read-only view (throws `UnsupportedOperationException` on write).


````java
import java.util.ArrayList;
import java.util.Arrays;
import java.util.Collections;
import java.util.List;

List<Integer> numbersList = new ArrayList<>(Arrays.asList(40, 10, 80, 20, 60, 20, 90));
System.out.println("Initial List: " + numbersList);

// 1. Min and Max
System.out.println("Min element: " + Collections.min(numbersList));
System.out.println("Max element: " + Collections.max(numbersList));

// 2. Frequency
System.out.println("Frequency of 20: " + Collections.frequency(numbersList, 20));

// 3. Reverse
Collections.reverse(numbersList);
System.out.println("Reversed List: " + numbersList);

// 4. Binary Search (Requires sorted list)
Collections.sort(numbersList);
int searchIdx = Collections.binarySearch(numbersList, 60);
System.out.println("Sorted List: " + numbersList);
System.out.println("Index of 60 via binarySearch: " + searchIdx);

// 5. Disjoint check
List<Integer> other = Arrays.asList(1, 2, 3);
System.out.println("Are lists disjoint (no elements in common)? " + Collections.disjoint(numbersList, other));

// 6. Unmodifiable Read-Only Wrapper
List<Integer> readOnlyList = Collections.unmodifiableList(numbersList);
System.out.println("Created unmodifiable list view of size: " + readOnlyList.size());
````

## 3. Generics and Wildcards (`<?>`, `<? extends T>`, `<? super T>`)
Generics provide **compile-time type safety** and eliminate explicit casting.
- **Generic Class & Generic Method**: Define type placeholders `<T>`.
- **Unbounded Wildcard (`<?>`)**: Accepts any type.
- **Upper-Bounded Wildcard (`<? extends Number>`)**: Accepts `Number` or any subclass (`Integer`, `Double`). **Read-safe (covariance)**.
- **Lower-Bounded Wildcard (`<? super Integer>`)**: Accepts `Integer` or any superclass (`Number`, `Object`). **Write-safe (contravariance)**.


````java
import java.util.List;
import java.util.Arrays;

class GenericHelper {
    // Generic method with type parameter T
    public static <T> void printElement(T element) {
        System.out.println("Type: " + element.getClass().getSimpleName() + " | Value: " + element);
    }

    // Upper-bounded wildcard: reads numbers (Integer, Double, etc.)
    public static double sumOfList(List<? extends Number> list) {
        double sum = 0.0;
        for (Number n : list) {
            sum += n.doubleValue();
        }
        return sum;
    }
}

GenericHelper.printElement(100);
GenericHelper.printElement("Hello Generics");

List<Integer> intList = Arrays.asList(10, 20, 30);
List<Double> doubleList = Arrays.asList(2.5, 4.5, 3.0);

System.out.println("Sum of Integers: " + GenericHelper.sumOfList(intList));
System.out.println("Sum of Doubles : " + GenericHelper.sumOfList(doubleList));
````

````java
import java.util.ArrayList;
import java.util.List;
import java.util.Arrays;

// 1. Defining a Custom Generic Class (Topic 81)
class Box<T> {
    private T content;

    public Box(T content) {
        this.content = content;
    }

    public T getContent() {
        return content;
    }

    public void setContent(T content) {
        this.content = content;
    }
}

Box<String> stringBox = new Box<>("CDAC Pune");
Box<Integer> intBox = new Box<>(1001);
System.out.println("Generic Box Content (String): " + stringBox.getContent());
System.out.println("Generic Box Content (Integer): " + intBox.getContent());

// 2. Comprehensive Wildcards Demonstration (Topic 82)
class WildcardDemo {
    // A. Unbounded Wildcard (<?>): Can read and print elements of any type
    public static void printAnyList(List<?> list) {
        System.out.print("Unbounded List [<?>]: ");
        for (Object item : list) {
            System.out.print(item + " ");
        }
        System.out.println();
    }

    // B. Lower-Bounded Wildcard (<? super Integer>): Safe for WRITING Integer objects
    public static void addIntegers(List<? super Integer> list) {
        list.add(100); // perfectly legal because list can accept Integer or higher
        list.add(200);
        System.out.println("Lower-bounded List after insertion: " + list);
    }
}

List<String> names = Arrays.asList("Alice", "Bob", "Charlie");
WildcardDemo.printAnyList(names); // Unbounded wildcard works on String list

List<Number> numList = new ArrayList<>();
WildcardDemo.addIntegers(numList); // Lower-bounded wildcard accepts List<Number> or List<Object>
````

## 4. Modern Date and Time API (`java.time`)
Introduced in Java 8 to replace legacy `java.util.Date` and `Calendar`:
- **Immutable and Thread-safe**.
- `LocalDate`: Date only (`yyyy-MM-dd`).
- `LocalTime`: Time only (`HH:mm:ss`).
- `LocalDateTime`: Combined date and time without timezone.
- `DateTimeFormatter`: Formatting to string and parsing from string.


````java
import java.time.LocalDate;
import java.time.LocalDateTime;
import java.time.format.DateTimeFormatter;

// 1. Current Date and Time
LocalDate today = LocalDate.now();
LocalDateTime now = LocalDateTime.now();

System.out.println("Current Date: " + today);
System.out.println("Current Date & Time: " + now);

// 2. Date manipulation (plus / minus)
LocalDate nextWeek = today.plusDays(7);
System.out.println("Date 7 days from now: " + nextWeek);

// 3. Formatting Date to custom String pattern
DateTimeFormatter formatter = DateTimeFormatter.ofPattern("dd-MMM-yyyy");
String formattedDate = today.format(formatter);
System.out.println("Formatted Date: " + formattedDate);

// 4. Parsing String to LocalDate
String dateInput = "15-Aug-2026";
LocalDate parsedDate = LocalDate.parse(dateInput, formatter);
System.out.println("Parsed LocalDate: " + parsedDate);
````
