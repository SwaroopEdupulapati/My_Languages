# Arrays in Java
An **array** is a fixed-size, homogeneous data structure that stores elements of the same data type in contiguous memory locations.

### Types of Arrays:
1. **1D Arrays (Single-Dimensional)**
2. **2D Arrays (Matrix / Multi-Dimensional)**
3. **Jagged Arrays (Ragged / Uneven Rows)**
4. **Utility Methods (`java.util.Arrays`)**


---
## 1. Single-Dimensional (1D) Arrays
- Declaration: `int[] arr = new int[size];` or `int[] arr = {val1, val2, ...};`
- Indexing: 0-based (`0` to `length - 1`).


```java
// Declaration and direct initialization
int[] numbers = {10, 20, 30, 40, 50};

// Access and update by index
System.out.println("First element: " + numbers[0]);
numbers[2] = 99; // update index 2
System.out.println("Updated index 2: " + numbers[2]);
System.out.println("Array length: " + numbers.length);

// Traversal using enhanced for-each loop
System.out.print("Array elements: ");
for (int num : numbers) {
    System.out.print(num + " ");
}
System.out.println();
```

---
## 2. Multi-Dimensional (2D Matrix) Arrays
- Stored as an "array of arrays" (`int[rows][columns]`).


```java
// 2x3 Matrix
int[][] matrix = {
    {1, 2, 3},
    {4, 5, 6}
};

System.out.println("Matrix element at [1][2]: " + matrix[1][2]); // 6

System.out.println("--- Matrix Traversal ---");
for (int r = 0; r < matrix.length; r++) {
    for (int c = 0; c < matrix[r].length; c++) {
        System.out.print(matrix[r][c] + " ");
    }
    System.out.println();
}
```

---
## 3. Jagged Arrays (Uneven Columns)
A multi-dimensional array where each row has a different number of columns.


```java
// Declare jagged array with 3 rows
int[][] jagged = new int[3][];
jagged[0] = new int[]{10, 20};       // row 0 has 2 columns
jagged[1] = new int[]{30, 40, 50};   // row 1 has 3 columns
jagged[2] = new int[]{60};           // row 2 has 1 column

System.out.println("--- Jagged Array Traversal ---");
for (int i = 0; i < jagged.length; i++) {
    System.out.print("Row " + i + ": ");
    for (int j = 0; j < jagged[i].length; j++) {
        System.out.print(jagged[i][j] + " ");
    }
    System.out.println();
}
```

---
## 4. Useful `java.util.Arrays` Methods
- `Arrays.toString(arr)`: Print 1D array as string
- `Arrays.deepToString(arr)`: Print multi-dimensional array
- `Arrays.sort(arr)`: Sort elements ascending
- `Arrays.binarySearch(arr, key)`: Find element index (requires sorted array)
- `Arrays.equals(arr1, arr2)`: Compare contents for equality
- `Arrays.fill(arr, val)`: Fill array with a fixed value
- `Arrays.copyOf(arr, newLength)`: Copy or resize array


```java
import java.util.Arrays;

int[] data = {40, 10, 50, 20, 30};

// Sort
Arrays.sort(data);
System.out.println("Sorted array: " + Arrays.toString(data));

// Binary Search (returns index)
int searchIdx = Arrays.binarySearch(data, 30);
System.out.println("Index of 30: " + searchIdx);

// Copying
int[] copy = Arrays.copyOf(data, 3);
System.out.println("Copied first 3 elements: " + Arrays.toString(copy));

// Comparison
System.out.println("Arrays equal? " + Arrays.equals(data, copy));

// Fill
int[] filledArr = new int[4];
Arrays.fill(filledArr, 7);
System.out.println("Filled array: " + Arrays.toString(filledArr));
```
