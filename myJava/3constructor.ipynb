# Constructors in Java
A **constructor** is a special block of code called when an instance of a class is created with the `new` keyword.

### Key Rules:
1. Constructor name must match the class name exactly.
2. It must **not** have an explicit return type (not even `void`).
3. Its main purpose is **initializing object state / instance variables**.


---
## 1. Non-Parameterized Constructor
A constructor that takes no arguments. It initializes instance variables with default or fixed values.


```java
class Student {
    String name;
    int rollNo;

    // Non-parameterized constructor
    Student() {
        name = "Unknown";
        rollNo = 0;
        System.out.println("Non-parameterized constructor called.");
    }
}

Student s1 = new Student();
System.out.println("Name: " + s1.name + ", Roll No: " + s1.rollNo);
```

---
## 2. Parameterized Constructor
A constructor that accepts parameters to initialize an object with custom values at creation time.


```java
class Book {
    String title;
    double price;

    // Parameterized constructor
    Book(String t, double p) {
        title = t;
        price = p;
    }
}

Book b1 = new Book("Java Programming", 499.0);
System.out.println("Title: " + b1.title + ", Price: " + b1.price);
```

---
## 3. Value Initialization Using `this` Keyword
When parameter names match instance variable names, `this.variable` is used to distinguish the instance variable from the parameter (shadowing resolution).


```java
class Employee {
    int id;
    String name;
    double salary;

    // Constructor using 'this' to initialize instance variables
    Employee(int id, String name, double salary) {
        this.id = id;             // this.id refers to the instance variable
        this.name = name;         // this.name refers to the instance variable
        this.salary = salary;     // this.salary refers to the instance variable
    }

    void display() {
        System.out.println("ID: " + this.id + ", Name: " + this.name + ", Salary: " + this.salary);
    }
}

Employee emp1 = new Employee(101, "Swaroop", 75000.0);
emp1.display();
```

## 4. Constructor Overloading and `this()`
`this()` is used to call one constructor from another within the same class (must be the first line).


```java
class Account {
    int accNo;
    double balance;

    // Default/Non-parameterized constructor
    Account() {
        this(1001, 500.0); // Calls the parameterized constructor
    }

    // Parameterized constructor
    Account(int accNo, double balance) {
        this.accNo = accNo;
        this.balance = balance;
    }

    void show() {
        System.out.println("Account: " + this.accNo + ", Balance: " + this.balance);
    }
}

Account a1 = new Account();
Account a2 = new Account(1002, 12500.0);
a1.show();
a2.show();
```

## 4.1 Constructor Chaining: Same-Class, Inheritance & Across Packages (Topic 40)
**Constructor Chaining** is the process of calling one constructor from another constructor in sequence:

1. **Within Same Class (`this()`)**: Delegates initialization to overloaded constructors. Must be the **first line** of the calling constructor.
2. **Across Class Hierarchy (`super()`)**: Child class constructor calls parent class constructor. If not explicitly written, Java automatically inserts `super()` (default parent constructor).
3. **Constructor Chaining Across Packages**:
   - When class `Child` in `package com.b` extends `Parent` in `package com.a`, the parent constructor called via `super()` **must be `public` or `protected`**.
   - A `default` (package-private) parent constructor is inaccessible across package boundaries, resulting in a compilation error.


```java
// Demonstrating Multi-Level Constructor Chaining
class GrandParent {
    // Protected constructor accessible across packages by subclasses
    protected GrandParent(String lineage) {
        System.out.println("[1. GrandParent Constructor] Lineage: " + lineage);
    }
}

class ParentClass extends GrandParent {
    // Chaining to superclass
    protected ParentClass(String lineage, String name) {
        super(lineage); // chains to GrandParent constructor
        System.out.println("[2. Parent Constructor] Name: " + name);
    }
}

class ChildClass extends ParentClass {
    int age;

    // Constructor 1: uses this() chaining
    public ChildClass(String lineage, String name) {
        this(lineage, name, 18); // chains within same class
        System.out.println("[3. Child Constructor 1 (default age applied)]");
    }

    // Constructor 2: uses super() chaining
    public ChildClass(String lineage, String name, int age) {
        super(lineage, name); // chains to Parent constructor
        this.age = age;
        System.out.println("[3. Child Constructor 2] Age initialized to: " + age);
    }
}

System.out.println("--- Initiating Chained Constructor Invocation ---");
ChildClass child = new ChildClass("Solaris Lineage", "Alex");
```

## 5. Records in Java (Modern Java 16+)
A **`record`** is a concise way to create immutable data carrier classes:
- Automatically generates:
  - `private final` fields for all components.
  - A canonical constructor.
  - Getters (named after the components without `get` prefix, e.g. `p.name()`).
  - `toString()`, `equals()`, and `hashCode()` methods.
- Eliminates boilerplate data-transfer object (DTO) code.


```java
// Defining a record (replaces 50+ lines of standard boilerplate class)
record StudentRecord(int id, String name, double marks) {
    // Compact constructor for validation (optional)
    public StudentRecord {
        if (marks < 0.0 || marks > 100.0) {
            throw new IllegalArgumentException("Invalid marks: " + marks);
        }
    }
}

StudentRecord s1 = new StudentRecord(101, "Swaroop", 94.5);
StudentRecord s2 = new StudentRecord(101, "Swaroop", 94.5);

// Component access methods (no 'get' prefix)
System.out.println("Record ID   : " + s1.id());
System.out.println("Record Name : " + s1.name());
System.out.println("Record Marks: " + s1.marks());

// Automatically generated toString()
System.out.println("toString()  : " + s1);

// Automatically generated equals() comparing data components
System.out.println("s1.equals(s2): " + s1.equals(s2)); // true!
```

## 6. Stack Memory vs Heap Memory
Understanding object memory allocation:
- **Stack Memory**: Stores primitive local variables and reference pointers. Allocated per thread; follows LIFO; deallocated automatically when method execution ends.
- **Heap Memory**: Stores actual runtime objects, instance variables, and arrays. Managed by the JVM Garbage Collector (GC).


```java
// Memory Demonstration:
// 'localVar' and reference 'studentRef' live on the STACK frame of this execution.
int localVar = 50; 

// The actual StudentRecord object payload lives in the HEAP memory.
StudentRecord studentRef = new StudentRecord(201, "Aarav", 88.0);

System.out.println("Stack holds reference pointer -> Heap holds actual object data: " + studentRef);
```

## 7. Difference Between Reference Variables and Static Variables (Topic 23)

| Feature | Reference Variable | Static Variable (Class Variable) |
| :--- | :--- | :--- |
| **Definition** | A variable that stores the memory address (reference) of an object in Heap | A variable declared with `static` that belongs to the class itself |
| **Memory Location** | Reference pointer is allocated in **Stack** (if local) or inside an object in **Heap** (if instance field) | Allocated in the **Metaspace / Method Area** at class loading time |
| **Number of Copies** | One copy per object instance created (or per local variable scope) | Only **one single shared copy** for the entire class, regardless of instances created |
| **Lifecycle** | Created when method executes (local) or object is instantiated; destroyed when out of scope or garbage collected | Created when class is loaded by ClassLoader; destroyed when class is unloaded or JVM shuts down |
| **Access Syntax** | Accessed via variable name directly or reference handle (`objRef.field`) | Accessed directly via Class Name (`ClassName.varName`) |
| **Default Value** | Instance reference variables default to `null` | Defaults according to its type (`0`, `0.0`, `null`, `false`) |
| **Modifications** | Modifying one object's reference does not alter other objects | Modifying a static variable reflects immediately across **all** instances |


```java
class VariableComparisonDemo {
    // 1. Static Variable: 1 single copy shared in Metaspace
    static int globalCounter = 0;

    // 2. Reference Variable (instance field): points to independent heap String per instance
    String instanceLabel;

    VariableComparisonDemo(String label) {
        this.instanceLabel = label;
        globalCounter++; // increments shared static variable
    }
}

// Reference variables on Stack pointing to distinct Heap objects
VariableComparisonDemo refA = new VariableComparisonDemo("Object-A");
VariableComparisonDemo refB = new VariableComparisonDemo("Object-B");

System.out.println("refA label (unique instance): " + refA.instanceLabel);
System.out.println("refB label (unique instance): " + refB.instanceLabel);
System.out.println("Shared Static Counter (both see 2): " + VariableComparisonDemo.globalCounter);
```
