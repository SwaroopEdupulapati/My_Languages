# LinkedList in Java (`java.util.LinkedList`)
A **LinkedList** is a doubly-linked list implementing both `List` and `Deque` interfaces, offering standard list access alongside full double-ended queue operations.

### Complete Methods Reference:
| Category | Methods |
| :--- | :--- |
| **List Insertion** | `add(e)`, `add(index, e)`, `addAll(c)`, `addAll(index, c)` |
| **Deque Insertion** | `addFirst(e)`, `addLast(e)`, `offer(e)`, `offerFirst(e)`, `offerLast(e)` |
| **Access / Inspect** | `get(index)`, `getFirst()`, `getLast()`, `peek()`, `peekFirst()`, `peekLast()` |
| **Removal** | `remove(index)`, `remove(Object)`, `removeFirst()`, `removeLast()`, `poll()`, `pollFirst()`, `pollLast()` |
| **Occurrence Removal** | `removeFirstOccurrence(Object)`, `removeLastOccurrence(Object)` |
| **Stack Operations** | `push(e)` (same as `addFirst`), `pop()` (same as `removeFirst`) |
| **Query & Traversal** | `contains(o)`, `indexOf(o)`, `lastIndexOf(o)`, `size()`, `descendingIterator()`, `clear()` |


```java
import java.util.LinkedList;
import java.util.Iterator;

LinkedList<String> list = new LinkedList<>();

// 1. Insertions (List + Deque)
list.add("B");
list.add("C");
list.addFirst("A");
list.addLast("D");
list.add(2, "X"); // insert at index 2
System.out.println("LinkedList: " + list);

// 2. Access methods
System.out.println("get(2): " + list.get(2));
System.out.println("getFirst(): " + list.getFirst() + ", peekFirst(): " + list.peekFirst());
System.out.println("getLast(): " + list.getLast() + ", peekLast(): " + list.peekLast());

// 3. Stack emulation methods directly on LinkedList
list.push("TOP"); // equivalent to addFirst()
System.out.println("After push('TOP'): " + list);
System.out.println("pop(): " + list.pop()); // equivalent to removeFirst()

// 4. Removals
list.removeFirst();
list.removeLast();
list.remove("X");
System.out.println("After removals: " + list);

// 5. Duplicate handling with occurrence removal
list.add("A");
list.add("B");
list.add("A");
System.out.println("With duplicates: " + list);
list.removeFirstOccurrence("A");
list.removeLastOccurrence("A");
System.out.println("After occurrence removals: " + list);

// 6. Descending Iterator (Reverse traversal)
System.out.print("Reverse traversal: ");
Iterator<String> revIt = list.descendingIterator();
while (revIt.hasNext()) {
    System.out.print(revIt.next() + " ");
}
System.out.println();

// 7. Clear
list.clear();
System.out.println("After clear(): " + list);
```
