# Queue and Deque in Java (`java.util.Queue`, `java.util.Deque`)
A **Queue** represents a **First-In, First-Out (FIFO)** sequence. Using `ArrayDeque` or `LinkedList`, it also supports double-ended operations (Deque).

### Complete Methods Comparison Table:
| Operation | Throws Exception | Returns Special Value (`false`/`null`) |
| :--- | :--- | :--- |
| **Insert (Tail)** | `add(e)` | `offer(e)` |
| **Remove (Head)** | `remove()` | `poll()` |
| **Examine (Head)** | `element()` | `peek()` |

### Double-Ended (Deque) Methods:
- `addFirst(e)` / `offerFirst(e)`, `addLast(e)` / `offerLast(e)`
- `removeFirst()` / `pollFirst()`, `removeLast()` / `pollLast()`
- `getFirst()` / `peekFirst()`, `getLast()` / `peekLast()`
- `removeFirstOccurrence(o)`, `removeLastOccurrence(o)`


```java
import java.util.Queue;
import java.util.Deque;
import java.util.ArrayDeque;
import java.util.LinkedList;

// 1. Standard FIFO Queue Operations
Queue<String> queue = new LinkedList<>();
queue.offer("Task 1");
queue.offer("Task 2");
queue.add("Task 3");
System.out.println("Queue: " + queue);

System.out.println("peek(): " + queue.peek());       // head inspection
System.out.println("element(): " + queue.element()); // head inspection (throws if empty)
System.out.println("poll(): " + queue.poll());       // remove head
System.out.println("remove(): " + queue.remove());   // remove head (throws if empty)
System.out.println("size(): " + queue.size() + ", isEmpty(): " + queue.isEmpty());

// 2. Double-Ended Queue (Deque) Operations with ArrayDeque
Deque<Integer> deque = new ArrayDeque<>();
deque.offerFirst(10); // insert front
deque.offerLast(20);  // insert rear
deque.offerFirst(5);  // insert front
deque.offerLast(30);  // insert rear
System.out.println("\nDeque: " + deque); // [5, 10, 20, 30]

System.out.println("peekFirst(): " + deque.peekFirst()); // 5
System.out.println("peekLast() : " + deque.peekLast());  // 30
System.out.println("pollFirst(): " + deque.pollFirst()); // removes 5
System.out.println("pollLast() : " + deque.pollLast());  // removes 30
System.out.println("Deque after polls: " + deque);

// 3. Clear
deque.clear();
System.out.println("Deque after clear(): " + deque);
```
